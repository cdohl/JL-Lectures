# 3.1 Circular motion

## Polar Coordinates

So far, we have described the position of a point using its Cartesian coordinates

$$
P=(x,y).
$$

For circular motion, it is often more convenient to describe the same point by

- its distance $r$ from the origin, and
- the angle $\varphi$ measured from the positive $x$-axis.

These are called **polar coordinates**.

From trigonometry,

$$
\boxed{x=r\cos\varphi}
$$

and

$$
\boxed{y=r\sin\varphi}.
$$

Therefore, the position vector can be written as

$$
\boxed{
\vec r=
\begin{pmatrix}
x\\
y
\end{pmatrix}
=
\begin{pmatrix}
r\cos\varphi\\
r\sin\varphi
\end{pmatrix}.
}
$$

Conversely, the distance from the origin is

$$
\boxed{
r=\sqrt{x^2+y^2}.
}
$$

Thus, $(x,y)$ and $(r,\varphi)$ are simply **two different ways of
describing the same point**.

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display


def polar_coordinates(r=1.0, phi=45):

    phirad = np.deg2rad(phi)

    x = r*np.cos(phirad)
    y = r*np.sin(phirad)

    fig, ax = plt.subplots(figsize=(6, 6))

    # coordinate axes
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)

    # position vector
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.08,
        head_length=0.12,
        linewidth=2
    )

    # x and y projections
    ax.plot([x, x], [0, y], 'k--', linewidth=1)
    ax.plot([0, x], [y, y], 'k--', linewidth=1)

    # point P
    ax.plot(x, y, 'ko')
    ax.text(x + 0.08, y + 0.08, r'$P$', fontsize=16)

    # position vector label
    ax.text(
        0.55*x - 0.08*np.sin(phirad),
        0.55*y + 0.08*np.cos(phirad),
        r'$\vec{r}$',
        fontsize=18
    )

    # component labels
    ax.text(
        x/2, -0.12,
        r'$x=r\cos\varphi$',
        ha='center'
    )

    ax.text(
        x + 0.12, y/2,
        r'$y=r\sin\varphi$',
        va='center'
    )

    # angle arc
    arc_radius = 0.45

    arc = patches.Arc(
        (0, 0),
        2*arc_radius,
        2*arc_radius,
        angle=0,
        theta1=0,
        theta2=phi,
        linewidth=2
    )

    ax.add_patch(arc)

    # angle label
    phi_mid = phirad/2

    ax.text(
        0.58*np.cos(phi_mid),
        0.58*np.sin(phi_mid),
        r'$\varphi$',
        fontsize=16,
        ha='center',
        va='center'
    )

    # numerical information
    ax.text(
        0.03, 0.96,
        rf'$r={r:.2f}$' '\n'
        rf'$\varphi={phi:.0f}^\circ$' '\n'
        rf'$x={x:.2f}$' '\n'
        rf'$y={y:.2f}$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.9
        )
    )

    ax.set_xlim(-2.2, 2.2)
    ax.set_ylim(-2.2, 2.2)

    ax.set_aspect('equal')

    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')

    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)


r_slider = widgets.FloatSlider(
    value=1.0,
    min=0.2,
    max=2.0,
    step=0.1,
    description=r'$r$:'
)

phi_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description=r'$\varphi$:'
)

out = widgets.interactive_output(
    polar_coordinates,
    {
        'r': r_slider,
        'phi': phi_slider
    }
)

display(
    widgets.HBox([r_slider, phi_slider]),
    out
)
```

## Velocity in Polar Coordinates

The position of a moving particle can be written as

$$
x(t)=r(t)\cos\varphi(t)
$$

and

$$
y(t)=r(t)\sin\varphi(t).
$$

Both $r$ and $\varphi$ may change with time.

The Cartesian components of the velocity are

$$
v_x=\frac{dx}{dt},
\qquad
v_y=\frac{dy}{dt}.
$$

We differentiate $x(t)$ using the product rule and the chain rule:

$$
\begin{aligned}
v_x
&=\frac{d}{dt}\left(r\cos\varphi\right)\\[4pt]
&=\dot r\cos\varphi
-r\dot\varphi\sin\varphi.
\end{aligned}
$$

Similarly,

$$
\begin{aligned}
v_y
&=\frac{d}{dt}\left(r\sin\varphi\right)\\[4pt]
&=\dot r\sin\varphi
+r\dot\varphi\cos\varphi.
\end{aligned}
$$

Thus,

$$
\boxed{
v_x=
\dot r\cos\varphi
-r\dot\varphi\sin\varphi
}
$$

and

$$
\boxed{
v_y=
\dot r\sin\varphi
+r\dot\varphi\cos\varphi.
}
$$

These equations contain **two different kinds of motion**.

### 1. Radial motion

The terms

$$
\dot r\cos\varphi,
\qquad
\dot r\sin\varphi
$$

describe motion directly **towards or away from the origin**.

The radial velocity is therefore

$$
\boxed{v_r=\dot r}.
$$

### 2. Tangential motion

The remaining terms are

$$
-r\dot\varphi\sin\varphi,
\qquad
r\dot\varphi\cos\varphi.
$$

They describe motion **perpendicular to the radial direction**.

The angular velocity is

$$
\boxed{\omega=\dot\varphi}
$$

and the corresponding tangential velocity is

$$
\boxed{v_\varphi=r\dot\varphi=r\omega}.
$$

The complete velocity can therefore be understood as the sum of a
**radial velocity** and a **tangential velocity**:

$$
\boxed{
\vec v=\vec v_r+\vec v_\varphi.
}
$$

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def velocity_polar(phi=45, rdot=0.4, phidot=0.6):

    r = 1.0
    phirad = np.deg2rad(phi)

    # Position
    x = r*np.cos(phirad)
    y = r*np.sin(phirad)

    # Radial velocity
    vrx = rdot*np.cos(phirad)
    vry = rdot*np.sin(phirad)

    # Tangential velocity
    vphi = r*phidot

    vpx = -vphi*np.sin(phirad)
    vpy =  vphi*np.cos(phirad)

    # Total velocity
    vx = vrx + vpx
    vy = vry + vpy

    fig, ax = plt.subplots(figsize=(7, 7))

    # Circle
    alpha = np.linspace(0, 2*np.pi, 300)

    ax.plot(
        r*np.cos(alpha),
        r*np.sin(alpha),
        linewidth=1,
        alpha=0.5
    )

    # Coordinate axes
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)

    # Position vector
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        0.5*x,
        0.5*y,
        r'$\vec r$',
        fontsize=17
    )

    # Radial velocity
    ax.arrow(
        x, y,
        vrx, vry,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        x + 0.55*vrx,
        y + 0.55*vry,
        r'$\vec v_r$',
        fontsize=16
    )

    # Tangential velocity
    ax.arrow(
        x, y,
        vpx, vpy,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        x + 0.55*vpx,
        y + 0.55*vpy,
        r'$\vec v_\varphi$',
        fontsize=16
    )

    # Total velocity
    ax.arrow(
        x, y,
        vx, vy,
        length_includes_head=True,
        head_width=0.06,
        head_length=0.09,
        linewidth=2.5
    )

    ax.text(
        x + 0.55*vx,
        y + 0.55*vy,
        r'$\vec v$',
        fontsize=18
    )

    # Parallelogram construction
    ax.plot(
        [x + vrx, x + vx],
        [y + vry, y + vy],
        'k--',
        alpha=0.5
    )

    ax.plot(
        [x + vpx, x + vx],
        [y + vpy, y + vy],
        'k--',
        alpha=0.5
    )

    # Information
    ax.text(
        0.03, 0.97,
        rf'$\varphi={phi:.0f}^\circ$' '\n'
        rf'$v_r=\dot r={rdot:.2f}$' '\n'
        rf'$\omega=\dot\varphi={phidot:.2f}$' '\n'
        rf'$v_\varphi=r\dot\varphi={vphi:.2f}$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.9
        )
    )

    ax.set_xlim(-2.0, 2.0)
    ax.set_ylim(-2.0, 2.0)

    ax.set_aspect('equal')

    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')

    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)


phi_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description=r'$\varphi$:'
)

rdot_slider = widgets.FloatSlider(
    value=0.4,
    min=-1,
    max=1,
    step=0.1,
    description=r'$\dot r$:'
)

phidot_slider = widgets.FloatSlider(
    value=0.6,
    min=-1,
    max=1,
    step=0.1,
    description=r'$\dot\varphi$:'
)

out = widgets.interactive_output(
    velocity_polar,
    {
        'phi': phi_slider,
        'rdot': rdot_slider,
        'phidot': phidot_slider
    }
)

display(
    widgets.VBox([
        phi_slider,
        rdot_slider,
        phidot_slider
    ]),
    out
)
```

## Circular Motion is Always Accelerated Motion

For circular motion, the radius is constant:

$$
r=R=\text{constant},
\qquad
\dot r=0.
$$

If the angular velocity is also constant,

$$
\dot\varphi=\omega=\text{constant},
$$

the position is

$$
x(t)=R\cos(\omega t)
$$

and

$$
y(t)=R\sin(\omega t).
$$

Differentiating gives the velocity:

$$
v_x=-R\omega\sin(\omega t)
$$

$$
v_y=R\omega\cos(\omega t).
$$

The speed is therefore

$$
|\vec v|
=
\sqrt{v_x^2+v_y^2}
=
R\omega
\sqrt{\sin^2(\omega t)+\cos^2(\omega t)}.
$$

Since

$$
\sin^2\varphi+\cos^2\varphi=1,
$$

we obtain

$$
\boxed{|\vec v|=R\omega=\text{constant}.}
$$

So the particle moves with **constant speed**.

However, the velocity is a vector. Its **direction changes continuously**.

To find the acceleration, differentiate the velocity once more:

$$
a_x
=
\frac{dv_x}{dt}
=
-R\omega^2\cos(\omega t)
$$

and

$$
a_y
=
\frac{dv_y}{dt}
=
-R\omega^2\sin(\omega t).
$$

Therefore,

$$
\vec a
=
-R\omega^2
\begin{pmatrix}
\cos(\omega t)\\
\sin(\omega t)
\end{pmatrix}.
$$

But the position vector is

$$
\vec r
=
R
\begin{pmatrix}
\cos(\omega t)\\
\sin(\omega t)
\end{pmatrix}.
$$

Thus,

$$
\boxed{
\vec a=-\omega^2\vec r.
}
$$

The minus sign means that the acceleration points in the direction
**opposite to the position vector** — towards the centre of the circle.

Its magnitude is

$$
\boxed{
a=R\omega^2=\frac{v^2}{R}.
}
$$

### Important conclusion

Even though the **speed is constant**, the **velocity is not constant**
because its direction changes.

$$
\boxed{\text{Uniform circular motion is accelerated motion. }}
$$

This instantaneous acceleration is termed *centripetal acceleration*.


```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def circular_motion(phi=45):

    R = 1.0
    omega = 1.0

    phirad = np.deg2rad(phi)

    # Position
    x = R*np.cos(phirad)
    y = R*np.sin(phirad)

    # Velocity
    vx = -R*omega*np.sin(phirad)
    vy =  R*omega*np.cos(phirad)

    # Acceleration
    axx = -R*omega**2*np.cos(phirad)
    ayy = -R*omega**2*np.sin(phirad)

    fig, ax = plt.subplots(figsize=(7, 7))

    # Circle
    alpha = np.linspace(0, 2*np.pi, 400)

    ax.plot(
        R*np.cos(alpha),
        R*np.sin(alpha),
        linewidth=1.5
    )

    # coordinate axes
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)

    # moving point
    ax.plot(x, y, 'ko', markersize=7)

    # position vector
    ax.arrow(
        0, 0, x, y,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        0.5*x,
        0.5*y,
        r'$\vec r$',
        fontsize=18
    )

    # velocity vector
    ax.arrow(
        x, y, 0.65*vx, 0.65*vy,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        x + 0.4*vx,
        y + 0.4*vy,
        r'$\vec v$',
        fontsize=18
    )

    # acceleration vector
    ax.arrow(
        x, y, 0.65*axx, 0.65*ayy,
        length_includes_head=True,
        head_width=0.05,
        head_length=0.08,
        linewidth=2
    )

    ax.text(
        x + 0.4*axx,
        y + 0.4*ayy,
        r'$\vec a$',
        fontsize=18
    )

    ax.text(
        0.03, 0.97,
        r'$|\vec v|=R\omega=\mathrm{constant}$' '\n'
        r'$\vec a=-\omega^2\vec r$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.9
        )
    )

    ax.set_xlim(-1.7, 1.7)
    ax.set_ylim(-1.7, 1.7)

    ax.set_aspect('equal')
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')

    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)


phi_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description=r'$\varphi$:'
)

out = widgets.interactive_output(
    circular_motion,
    {'phi': phi_slider}
)

display(phi_slider, out)
```

## Period and Frequency of Circular Motion

For uniform circular motion, the particle returns to the same position
after one complete revolution.

The time required for **one complete revolution** is called the
**period** $T$.

$$
\boxed{T=\text{time for one revolution}}
$$

The unit of the period is seconds:

$$
[T]=\mathrm{s}.
$$

---

### Frequency

Instead of asking how long one revolution takes, we can ask:

> How many revolutions occur per second?

This is the **frequency** $f$.

If one revolution takes a time $T$, then

$$
\boxed{
f=\frac{1}{T}
}
$$

or equivalently

$$
\boxed{
T=\frac{1}{f}.
}
$$

The unit of frequency is

$$
[f]=\mathrm{s^{-1}}=\mathrm{Hz}.
$$

Thus,

$$
1\,\mathrm{Hz}=1\text{ revolution per second}.
$$

For example, if a particle completes one revolution every $0.5\,\mathrm{s}$,

$$
T=0.5\,\mathrm{s}
$$

and therefore

$$
f=\frac{1}{0.5\,\mathrm{s}}
=2\,\mathrm{Hz}.
$$

The particle completes two revolutions every second.

---

### Connection to angular velocity

During one complete revolution, the angle changes by

$$
\Delta\varphi=2\pi.
$$

If one revolution takes the time $T$, the angular velocity is therefore

$$
\omega
=
\frac{\Delta\varphi}{\Delta t}
=
\frac{2\pi}{T}.
$$

Since

$$
f=\frac{1}{T},
$$

we obtain the important relation

$$
\boxed{
\omega=2\pi f
}
$$

and therefore

$$
\boxed{
f=\frac{\omega}{2\pi}.
}
$$

The unit of angular velocity is

$$
[\omega]=\mathrm{rad\,s^{-1}}.
$$

Be careful not to confuse $f$ and $\omega$:

$$
\boxed{
f=\text{revolutions per second}
}
$$

whereas

$$
\boxed{
\omega=\text{angle travelled per second}.
}
$$

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display


def rotating_ball(frame=0, f=1.0):

    R = 1.0

    # Each frame corresponds to this amount of real time
    dt = 0.02
    t = frame * dt

    # Period and angular velocity
    T = 1/f
    omega = 2*np.pi*f

    # Angular position
    phi = omega*t

    # Position
    x = R*np.cos(phi)
    y = R*np.sin(phi)

    # Tangential velocity
    vx = -R*omega*np.sin(phi)
    vy =  R*omega*np.cos(phi)

    fig, ax = plt.subplots(figsize=(6, 6))

    # Circle
    alpha = np.linspace(0, 2*np.pi, 400)
    ax.plot(
        R*np.cos(alpha),
        R*np.sin(alpha),
        linewidth=1.5
    )

    # Coordinate axes
    ax.axhline(0, linewidth=1)
    ax.axvline(0, linewidth=1)

    # Radius
    ax.plot(
        [0, x],
        [0, y],
        linewidth=2
    )

    # Rotating ball
    ax.plot(
        x, y,
        'o',
        markersize=14
    )

    # Velocity vector
    # normalized length for display, otherwise it becomes too long
    vscale = 0.22

    ax.arrow(
        x, y,
        vscale*vx,
        vscale*vy,
        length_includes_head=True,
        head_width=0.06,
        head_length=0.09,
        linewidth=2
    )

    ax.text(
        x + 0.55*vscale*vx,
        y + 0.55*vscale*vy,
        r'$\vec v$',
        fontsize=16
    )

    # Information
    ax.text(
        0.03, 0.97,
        rf'$f={f:.2f}\ \mathrm{{Hz}}$' '\n'
        rf'$T={T:.2f}\ \mathrm{{s}}$' '\n'
        rf'$\omega=2\pi f={omega:.2f}\ \mathrm{{rad/s}}$' '\n'
        rf'$v=R\omega={R*omega:.2f}$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(
            boxstyle='round',
            facecolor='white',
            alpha=0.9
        )
    )

    ax.set_xlim(-1.7, 1.7)
    ax.set_ylim(-1.7, 1.7)

    ax.set_aspect('equal')
    ax.set_xlabel('$x$')
    ax.set_ylabel('$y$')
    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)


# Frequency slider
f_slider = widgets.FloatSlider(
    value=0.5,
    min=0.1,
    max=2.0,
    step=0.1,
    description=r'$f$ (Hz):',
    continuous_update=True
)

# Hidden frame slider
frame_slider = widgets.IntSlider(
    value=0,
    min=0,
    max=10000,
    step=1,
    layout=widgets.Layout(display='none')
)

# Play button
play = widgets.Play(
    value=0,
    min=0,
    max=10000,
    step=1,
    interval=20,
    description='Play'
)

# Link play button to frame number
widgets.jslink(
    (play, 'value'),
    (frame_slider, 'value')
)

# Interactive output
out = widgets.interactive_output(
    rotating_ball,
    {
        'frame': frame_slider,
        'f': f_slider
    }
)

display(
    widgets.HBox([
        play,
        f_slider
    ]),
    out
)
```
