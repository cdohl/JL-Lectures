## 10. Programming: sliding motion and the energy balance
A block of mass $m=1.0\,\mathrm{kg}$ is released from rest on a long incline at $\theta=30^\circ$. Take $\mu_s=0.40$ and initially $\mu_k=0.20$. The incline is long enough for the whole simulated motion. The coordinate $s$ is positive downhill, with $s=0$ at release. Set gravitational potential energy to zero at release, so

$$U(s)=-mgs\sin\theta.$$

The program below uses small time steps. First it updates the velocity, then the position using the updated velocity. The block slides downhill throughout the simulated interval.

**a)** Draw the free-body diagram. Check that gravity overcomes the maximum static friction. Derive the downhill acceleration used in the program.

**b)** Run the program and describe the position and speed graphs. Explain the signs of $U$ and the work done by friction.

**c)** Compare $E_{\rm mech}=K+U$ with $E_{\rm mech}+E_{\rm diss}$, where $E_{\rm diss}$ is the cumulative mechanical energy transferred into internal energy by sliding friction. Which quantity should be constant in the exact physical model, and why?

**d)** Repeat with $\mu_k=0$, $0.10$, and $0.30$, keeping $\mu_s=0.40$. Predict first how the final speed and mechanical energy will change; then compare the results. Check that downhill motion remains consistent in all cases.

**e)** For $\mu_k=0.20$, reduce the time step from $0.010\,\mathrm s$ to $0.005\,\mathrm s$. Compare the printed energy-balance error. Explain why a small numerical drift does not imply a failure of energy conservation in the physical system.

Submit the plots for the initial case, a small table comparing the four friction coefficients, and the time-step comparison. You do not need to rewrite the numerical method.

```python
import numpy as np
import matplotlib.pyplot as plt

# SI units: kg, m, s, N, J
m = 1.0
g = 10.0
theta = np.deg2rad(30.0)
mu_k = 0.20
dt = 0.010
t_end = 1.0

steps = int(round(t_end / dt))
t = np.linspace(0.0, t_end, steps + 1)
dt = t[1] - t[0]
s = np.zeros_like(t)
v = np.zeros_like(t)

normal = m * g * np.cos(theta)
friction = mu_k * normal
acceleration = (m * g * np.sin(theta) - friction) / m
assert acceleration > 0, "This model requires downhill sliding."

for i in range(steps):
    v[i + 1] = v[i] + acceleration * dt
    s[i + 1] = s[i] + v[i + 1] * dt

K = 0.5 * m * v**2
U = -m * g * s * np.sin(theta)
E_mech = K + U
E_diss = friction * s
E_balance = E_mech + E_diss

fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
axes[0].plot(t, s)
axes[0].set(xlabel="time (s)", ylabel="position s (m)")
axes[1].plot(t, v)
axes[1].set(xlabel="time (s)", ylabel="speed (m/s)")
for ax in axes:
    ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

plt.figure(figsize=(7, 4))
for values, label in [(K, "K"), (U, "U"), (E_mech, "K + U"),
                      (E_diss, "E_diss"), (E_balance, "K + U + E_diss")]:
    plt.plot(t, values, label=label)
plt.xlabel("time (s)")
plt.ylabel("energy (J)")
plt.legend()
plt.grid(alpha=0.3)
plt.tight_layout()
plt.show()

print(f"Final speed: {v[-1]:.3f} m/s")
print(f"Final mechanical energy: {E_mech[-1]:.4f} J")
print(f"Final dissipated energy: {E_diss[-1]:.4f} J")
print(f"Energy-balance error: {E_balance[-1]-E_balance[0]:.6f} J")
```
