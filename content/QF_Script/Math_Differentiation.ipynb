## Mathematics for Biology Students

In mechanics we have already used derivatives to describe velocity and acceleration:
$$v(t)=\frac{dx}{dt},\qquad a(t)=\frac{dv}{dt}=\frac{d^2x}{dt^2}.$$

Now we learn **how to calculate derivatives**. A derivative describes a rate of change and geometrically gives the **slope of the tangent** to a graph.

## 1. From secant to tangent

For two points, the slope is
$$m=\frac{\Delta y}{\Delta x}.$$

As the second point approaches the first, the secant approaches the tangent:
$$\boxed{f'(x)=\lim_{\Delta x\to0}\frac{f(x+\Delta x)-f(x)}{\Delta x}}.$$

Reduce $\Delta x$ below and watch the secant approach the tangent.

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def tangent_demo(x0=1.0, dx=1.5):
    f=lambda x:x**2
    x=np.linspace(-3,3,400)
    x1=x0+dx
    ms=(f(x1)-f(x0))/dx
    mt=2*x0
    fig,ax=plt.subplots(figsize=(8,6))
    ax.plot(x,f(x),label=r'$f(x)=x^2$')
    ax.plot([x0,x1],[f(x0),f(x1)],'ko')
    ax.plot(x,f(x0)+ms*(x-x0),'--',label=rf'secant slope = {ms:.2f}')
    ax.plot(x,f(x0)+mt*(x-x0),label=rf'tangent slope = {mt:.2f}')
    ax.set(xlim=(-3,3),ylim=(-1,9),xlabel='$x$',ylabel='$f(x)$')
    ax.grid(alpha=.25); ax.legend()
    ax.set_title(r'Secant $\rightarrow$ tangent as $\Delta x\rightarrow0$')
    plt.show(); plt.close(fig)

x0=widgets.FloatSlider(value=1,min=-2,max=2,step=.1,description=r'$x_0$:')
dx=widgets.FloatSlider(value=1.5,min=.05,max=2,step=.05,description=r'$\Delta x$:')
out=widgets.interactive_output(tangent_demo,{'x0':x0,'dx':dx})
display(widgets.VBox([x0,dx]),out)
```

## 2. Deriving $f(x)=x^2$ once from the definition

$$f'(x)=\lim_{\Delta x\to0}\frac{(x+\Delta x)^2-x^2}{\Delta x}$$
$$=\lim_{\Delta x\to0}\frac{2x\Delta x+(\Delta x)^2}{\Delta x}$$
$$=\lim_{\Delta x\to0}(2x+\Delta x)=\boxed{2x}.$$

Fortunately, we do not need to do this every time. We use **differentiation rules**.

## 3. Notation

If $y=f(x)$, all of these denote a derivative:
$$\boxed{f'(x)=y'=\frac{df}{dx}=\frac{dy}{dx}}.$$

For a function of time we similarly write $dx/dt$.

# 4. The power rule

$$\boxed{\frac{d}{dx}x^n=nx^{n-1}}$$

**Recipe:** bring the exponent down in front and reduce it by one.

$$x^2\to2x,\qquad x^3\to3x^2,\qquad x^5\to5x^4.$$

The rule also works for negative and fractional powers:
$$\frac1x=x^{-1}\to-\frac1{x^2},$$
$$\sqrt{x}=x^{1/2}\to\frac{1}{2\sqrt{x}}.$$

# 5. Constants, factors, sums and differences

A constant does not change:
$$\boxed{\frac{d}{dx}C=0}.$$

A constant factor stays in front:
$$\boxed{\frac{d}{dx}[Cf(x)]=Cf'(x)}.$$

Differentiate sums and differences term by term:
$$\boxed{(f+g)'=f'+g',\qquad(f-g)'=f'-g'}.$$

Example:
$$f(x)=3x^4-2x^2+7x-5$$
$$\boxed{f'(x)=12x^3-4x+7}.$$

# 6. The product rule

For $h(x)=f(x)g(x)$,
$$\boxed{(fg)'=f'g+fg'}.$$

Remember:

**derivative of first × second + first × derivative of second**

Example:
$$h(x)=x^2\sin x$$
$$\boxed{h'(x)=2x\sin x+x^2\cos x}.$$

# 7. The quotient rule

For $h(x)=f(x)/g(x)$,
$$\boxed{\left(\frac fg\right)'=\frac{f'g-fg'}{g^2}}.$$

Example:
$$h(x)=\frac{x^2}{x+1}$$
$$h'(x)=\frac{2x(x+1)-x^2}{(x+1)^2}
=\boxed{\frac{x^2+2x}{(x+1)^2}}.$$

# 8. Important functions

| $f(x)$ | $f'(x)$ |
|---|---|
| $C$ | $0$ |
| $x^n$ | $nx^{n-1}$ |
| $e^x$ | $e^x$ |
| $\ln x$ | $1/x$ |
| $\sin x$ | $\cos x$ |
| $\cos x$ | $-\sin x$ |

The exponential and logarithmic functions are especially important in biology.

# 9. The chain rule

Use the chain rule when **one function is inside another**:
$$\boxed{\frac{d}{dx}f(g(x))=f'(g(x))\,g'(x)}.$$

Example:
$$f(x)=(3x+2)^4.$$

Differentiate the outside, keep the inside:
$$4(3x+2)^3,$$
then multiply by the derivative of the inside, $3$:
$$\boxed{f'(x)=12(3x+2)^3}.$$

**Recipe:** differentiate the outside, keep the inside, multiply by the derivative of the inside.

# 10. Biological example: exponential growth

A simple population model is
$$N(t)=N_0e^{kt}.$$

Using the chain rule,
$$\frac{dN}{dt}=N_0ke^{kt}.$$

Since $N(t)=N_0e^{kt}$,
$$\boxed{\frac{dN}{dt}=kN}.$$

Thus the **rate of population growth is proportional to the population itself**. The same mathematics appears in bacterial growth, decay processes, and simple models of drug elimination.

# 11. Differentiation cheat sheet

### Basic derivatives
$$\frac{d}{dx}C=0,\qquad
\frac{d}{dx}x^n=nx^{n-1}$$
$$\frac{d}{dx}e^x=e^x,\qquad
\frac{d}{dx}\ln x=\frac1x$$
$$\frac{d}{dx}\sin x=\cos x,\qquad
\frac{d}{dx}\cos x=-\sin x$$

### Rules
$$ (Cf)'=Cf' $$
$$ (f+g)'=f'+g' $$
$$ (fg)'=f'g+fg' $$
$$ \left(\frac fg\right)'=\frac{f'g-fg'}{g^2} $$
$$ \frac{d}{dx}f(g(x))=f'(g(x))g'(x) $$
