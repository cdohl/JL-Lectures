## 💻 10. Programming — integrating a force to predict motion

Return to the increasing-force cart in Problem 4. The program below uses Newton's second law and **numerical integration** to calculate velocity and position. This reverses the numerical differentiation used in Homework 3.

For samples separated by $\Delta t$, the trapezoidal rule gives

$$v_{i+1}=v_i+\frac{a_i+a_{i+1}}{2}\Delta t,$$

$$x_{i+1}=x_i+\frac{v_i+v_{i+1}}{2}\Delta t.$$

The code adds these increments using `np.cumsum`. The curves are predictions of the model, not experimental measurements.

### Tasks

**a)** Run the program. Describe the shapes of the acceleration, velocity and position graphs. Which of these quantities change linearly with time?

**b)** Compare the final velocity and position with your analytical answers in Problem 4. Are the results consistent?

**c)** Double the mass, keeping the force function unchanged. What happens to the final velocity and displacement? Explain using $F=ma$.

**d)** Restore the original mass and double $\beta$. Predict the changes before running the program.

**e)** Restore the original parameters and reduce the number of samples from 201 to 11. Compare the final position error with the finer calculation. Why can the velocity integral be exact to numerical precision for this linear acceleration, while the position integral still has a small approximation error?

> Trapezoids represent the area under a straight line exactly. This need not hold for a curved graph.

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters: use SI units throughout.
mass = 0.50             # kg
beta = 0.30             # N/s
duration = 2.0          # s
samples = 201
v0 = 0.0               # m/s
x0 = 0.0               # m

t = np.linspace(0.0, duration, samples)
dt = t[1] - t[0]
force = beta * t
acceleration = force / mass

velocity = np.empty_like(t)
position = np.empty_like(t)
velocity[0] = v0
position[0] = x0

# Cumulative trapezoidal integration, without extra packages.
velocity[1:] = v0 + np.cumsum(
    0.5 * (acceleration[:-1] + acceleration[1:]) * dt
)
position[1:] = x0 + np.cumsum(
    0.5 * (velocity[:-1] + velocity[1:]) * dt
)

fig, axes = plt.subplots(1, 3, figsize=(12, 3.5))
for ax, values, label in zip(
    axes,
    [acceleration, velocity, position],
    ["Acceleration (m/s^2)", "Velocity (m/s)", "Position (m)"]
):
    ax.plot(t, values)
    ax.set_xlabel("Time (s)")
    ax.set_ylabel(label)
    ax.grid(True)
plt.tight_layout()
plt.show()

print(f"Final velocity: {velocity[-1]:.6f} m/s")
print(f"Final position: {position[-1]:.6f} m")
```
