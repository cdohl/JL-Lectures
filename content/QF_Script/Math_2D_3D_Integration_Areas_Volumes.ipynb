# 2D and 3D Integration: Areas and Volumes

In one dimension, integration can be interpreted as adding small lengths.  
In two dimensions, we add small **area elements**,

$$
A=\iint_D dA,
$$

and in three dimensions, small **volume elements**,

$$
V=\iiint_\Omega dV.
$$

The main challenge is therefore to

1. describe the region of integration,
2. choose suitable coordinates,
3. determine the correct differential area or volume element,
4. specify the integration limits.

We begin with Cartesian coordinates and then introduce polar, cylindrical, and spherical coordinates.

## 1. Area in Cartesian coordinates

In Cartesian coordinates, a small rectangle with side lengths \(dx\) and \(dy\) has area

$$
dA = dx\,dy.
$$

Therefore,

$$
A=\iint_D dx\,dy.
$$

### Example 1: Rectangle

Consider

$$
0\le x\le a,\qquad 0\le y\le b.
$$

Then

$$
A=\int_0^a\int_0^b dy\,dx.
$$

The inner integral gives

$$
\int_0^b dy=b,
$$

and therefore

$$
A=\int_0^a b\,dx=ab.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

a, b = 4, 2

fig, ax = plt.subplots(figsize=(6, 4))
ax.fill([0,a,a,0], [0,0,b,b], alpha=0.25)
ax.plot([0,a,a,0,0], [0,0,b,b,0])
ax.set_xlim(-0.5, 4.5)
ax.set_ylim(-0.5, 2.5)
ax.set_aspect('equal')
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_title(r'Rectangle: $0\leq x\leq a,\;0\leq y\leq b$')
plt.show()
```

### Example 2: Area below a curve

Find the area bounded by

$$
y=x,\qquad y=0,\qquad 0\le x\le 2.
$$

For each value of \(x\),

$$
0\le y\le x.
$$

Thus,

$$
A=\int_0^2\int_0^x dy\,dx
 =\int_0^2 x\,dx
 =2.
$$

Notice how the upper integration limit of \(y\) depends on \(x\).

```python
x = np.linspace(0, 2, 300)

fig, ax = plt.subplots(figsize=(6, 4))
ax.fill_between(x, 0, x, alpha=0.25)
ax.plot(x, x, label='$y=x$')
ax.axhline(0, linewidth=0.8)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_aspect('equal')
ax.set_title('Region of integration')
ax.legend()
plt.show()
```

## 2. Polar coordinates

Circular regions are often cumbersome in Cartesian coordinates. For example, a disk of radius \(R\) is described by

$$
-R\le x\le R,
\qquad
-\sqrt{R^2-x^2}\le y\le\sqrt{R^2-x^2}.
$$

A much more natural description uses **polar coordinates**:

$$
x=r\cos\varphi,\qquad
y=r\sin\varphi.
$$

Here,

$$
0\le r\le R,\qquad 0\le\varphi\le2\pi.
$$

However, the area element is **not** \(dr\,d\varphi\).

### The polar area element

A small polar element is approximately a rectangle.

Its radial side has length

$$
dr,
$$

while its tangential side has length

$$
r\,d\varphi.
$$

Therefore,

$$
\boxed{dA=r\,dr\,d\varphi}.
$$

The factor \(r\) is the **Jacobian** of the transformation from Cartesian to polar coordinates.

```python
R = 2.0
r0 = 1.3
dr = 0.35
phi0 = 0.55
dphi = 0.35

fig, ax = plt.subplots(figsize=(6, 6))

phi = np.linspace(0, 2*np.pi, 400)
ax.plot(R*np.cos(phi), R*np.sin(phi))

phiw = np.linspace(phi0, phi0+dphi, 80)
xx = np.r_[r0*np.cos(phiw), (r0+dr)*np.cos(phiw[::-1])]
yy = np.r_[r0*np.sin(phiw), (r0+dr)*np.sin(phiw[::-1])]
ax.fill(xx, yy, alpha=0.35)

ax.plot([0, R*np.cos(phi0)], [0, R*np.sin(phi0)], '--', linewidth=0.8)
ax.plot([0, R*np.cos(phi0+dphi)], [0, R*np.sin(phi0+dphi)], '--', linewidth=0.8)

ax.text(0.9, 0.25, r'$r$', fontsize=12)
ax.text(1.35, 1.1, r'$r\,d\varphi$', fontsize=12)
ax.text(1.55, 0.9, r'$dr$', fontsize=12)

ax.set_aspect('equal')
ax.set_xlim(-2.3, 2.3)
ax.set_ylim(-2.3, 2.3)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_title(r'Polar area element: $dA=r\,dr\,d\varphi$')
plt.show()
```

### Example 3: Area of a disk

For a disk of radius \(R\),

$$
0\le r\le R,\qquad
0\le\varphi\le2\pi.
$$

Hence

$$
A
=
\int_0^{2\pi}\int_0^R r\,dr\,d\varphi.
$$

First integrate over \(r\):

$$
\int_0^R r\,dr=\frac{R^2}{2}.
$$

Then,

$$
A
=
\int_0^{2\pi}\frac{R^2}{2}\,d\varphi
=
\pi R^2.
$$

Thus the familiar formula for the area of a circle follows directly from integration.

# 3. Volume in Cartesian coordinates

A small Cartesian box has side lengths

$$
dx,\qquad dy,\qquad dz,
$$

so its volume is

$$
\boxed{dV=dx\,dy\,dz}.
$$

Therefore,

$$
V=\iiint_\Omega dx\,dy\,dz.
$$

### Example 4: Rectangular box

For

$$
0\le x\le a,\qquad
0\le y\le b,\qquad
0\le z\le c,
$$

we obtain

$$
V=
\int_0^a\int_0^b\int_0^c dz\,dy\,dx
=abc.
$$

```python
from mpl_toolkits.mplot3d.art3d import Poly3DCollection

a, b, c = 3, 2, 1.5

verts = np.array([
    [0,0,0], [a,0,0], [a,b,0], [0,b,0],
    [0,0,c], [a,0,c], [a,b,c], [0,b,c]
])

faces = [
    [verts[i] for i in [0,1,2,3]],
    [verts[i] for i in [4,5,6,7]],
    [verts[i] for i in [0,1,5,4]],
    [verts[i] for i in [1,2,6,5]],
    [verts[i] for i in [2,3,7,6]],
    [verts[i] for i in [3,0,4,7]]
]

fig = plt.figure(figsize=(7,5))
ax = fig.add_subplot(111, projection='3d')
poly = Poly3DCollection(faces, alpha=0.18, edgecolor='black')
ax.add_collection3d(poly)
ax.set_xlim(0,a)
ax.set_ylim(0,b)
ax.set_zlim(0,c)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$z$')
ax.set_title('Rectangular box')
plt.show()
```

## 4. Cylindrical coordinates

A cylinder suggests extending polar coordinates by adding the vertical coordinate \(z\):

$$
x=r\cos\varphi,\qquad
y=r\sin\varphi,\qquad
z=z.
$$

The volume element is obtained by multiplying the polar area element by \(dz\):

$$
dV=dA\,dz.
$$

Since

$$
dA=r\,dr\,d\varphi,
$$

we have

$$
\boxed{dV=r\,dr\,d\varphi\,dz}.
$$

### Example 5: Volume of a cylinder

Consider a cylinder of radius \(R\) and height \(H\).

Its limits are

$$
0\le r\le R,
\qquad
0\le\varphi\le2\pi,
\qquad
0\le z\le H.
$$

Therefore,

$$
V=
\int_0^H
\int_0^{2\pi}
\int_0^R
r\,dr\,d\varphi\,dz.
$$

Evaluating the integrals,

$$
V
=
H(2\pi)\frac{R^2}{2}
=
\boxed{\pi R^2H}.
$$

```python
R, H = 2, 3

phi = np.linspace(0, 2*np.pi, 80)
z = np.linspace(0, H, 40)
Phi, Z = np.meshgrid(phi, z)
X = R*np.cos(Phi)
Y = R*np.sin(Phi)

fig = plt.figure(figsize=(7,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.22)

rr = np.linspace(0, R, 30)
pp = np.linspace(0, 2*np.pi, 80)
RR, PP = np.meshgrid(rr, pp)
ax.plot_surface(RR*np.cos(PP), RR*np.sin(PP), 0*RR, alpha=0.15)
ax.plot_surface(RR*np.cos(PP), RR*np.sin(PP), H+0*RR, alpha=0.15)

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$z$')
ax.set_title(r'Cylinder: $V=\pi R^2H$')
ax.set_box_aspect((1,1,1.2))
plt.show()
```

# 5. Spherical coordinates

For a sphere, cylindrical coordinates are possible, but spherical coordinates are even more natural.

We use

$$
x=r\sin\theta\cos\varphi,
$$

$$
y=r\sin\theta\sin\varphi,
$$

$$
z=r\cos\theta,
$$

where

- \(r\) is the distance from the origin,
- \(\theta\) is the polar angle measured from the positive \(z\)-axis,
- \(\varphi\) is the azimuthal angle around the \(z\)-axis.

For a complete sphere,

$$
0\le r\le R,
\qquad
0\le\theta\le\pi,
\qquad
0\le\varphi\le2\pi.
$$

## The spherical volume element

Consider a very small element in spherical coordinates.

Its three approximate side lengths are

$$
dr,
$$

$$
r\,d\theta,
$$

and

$$
r\sin\theta\,d\varphi.
$$

Multiplying them gives

$$
dV
=
dr\,(r\,d\theta)\,(r\sin\theta\,d\varphi).
$$

Thus,

$$
\boxed{dV=r^2\sin\theta\,dr\,d\theta\,d\varphi}.
$$

The Jacobian is therefore

$$
\boxed{J=r^2\sin\theta}.
$$

### Example 6: Volume of a sphere

For a sphere of radius \(R\),

$$
V
=
\int_0^{2\pi}
\int_0^\pi
\int_0^R
r^2\sin\theta
\,dr\,d\theta\,d\varphi.
$$

The three integrations separate:

$$
V
=
\left(\int_0^R r^2\,dr\right)
\left(\int_0^\pi\sin\theta\,d\theta\right)
\left(\int_0^{2\pi}d\varphi\right).
$$

These give

$$
\frac{R^3}{3},
\qquad
2,
\qquad
2\pi.
$$

Therefore,

$$
\boxed{V=\frac{4}{3}\pi R^3}.
$$

```python
R = 2

u = np.linspace(0, 2*np.pi, 100)
v = np.linspace(0, np.pi, 60)
U, V = np.meshgrid(u, v)

X = R*np.sin(V)*np.cos(U)
Y = R*np.sin(V)*np.sin(U)
Z = R*np.cos(V)

fig = plt.figure(figsize=(7,6))
ax = fig.add_subplot(111, projection='3d')
ax.plot_surface(X, Y, Z, alpha=0.25)
ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$z$')
ax.set_title(r'Sphere: $V=\frac{4}{3}\pi R^3$')
ax.set_box_aspect((1,1,1))
plt.show()
```

# 6. Choosing coordinates

The shape of the object often tells us which coordinate system is most convenient.

| Geometry | Coordinates | Differential element |
|---|---|---|
| Rectangle / general \(x,y\) region | Cartesian | \(dA=dx\,dy\) |
| Disk / circular region | Polar | \(dA=r\,dr\,d\varphi\) |
| Box / general \(x,y,z\) region | Cartesian | \(dV=dx\,dy\,dz\) |
| Cylinder / axial symmetry | Cylindrical | \(dV=r\,dr\,d\varphi\,dz\) |
| Sphere / radial symmetry | Spherical | \(dV=r^2\sin\theta\,dr\,d\theta\,d\varphi\) |

The important idea is:

> **Choose coordinates that make the boundaries of the object simple.**

## Summary

For areas,

$$
\boxed{dA=dx\,dy}
$$

in Cartesian coordinates and

$$
\boxed{dA=r\,dr\,d\varphi}
$$

in polar coordinates.

For volumes,

$$
\boxed{dV=dx\,dy\,dz}
$$

in Cartesian coordinates,

$$
\boxed{dV=r\,dr\,d\varphi\,dz}
$$

in cylindrical coordinates, and

$$
\boxed{dV=r^2\sin\theta\,dr\,d\theta\,d\varphi}
$$

in spherical coordinates.

The additional factors \(r\) and \(r^2\sin\theta\) account for the fact that the physical size of an angular interval changes with position.


# 6. Application: Moment of inertia of a cylinder

Volume integration can also be used to calculate physical quantities.

Consider a homogeneous solid cylinder of radius $R$, height $H$, and mass $M$, rotating around its axis of symmetry (the $z$-axis).

A small mass element $dm$ at a distance $r$ from the axis contributes

$$
dI = r^2\,dm
$$

to the moment of inertia.

Therefore, the total moment of inertia is

$$
I = \iiint_\Omega r^2\,dm.
$$

For a homogeneous material with constant mass density $\rho$,

$$
dm = \rho\,dV.
$$

In cylindrical coordinates,

$$
dV = r\,dr\,d\varphi\,dz.
$$

Thus,

$$
dm = \rho r\,dr\,d\varphi\,dz.
$$



## Setting up the integral

For the cylinder,

$$
0\le r\le R,
$$

$$
0\le\varphi\le2\pi,
$$

and

$$
0\le z\le H.
$$

The moment of inertia is therefore

$$
I
=
\int_0^H
\int_0^{2\pi}
\int_0^R
r^2\,\rho r\,dr\,d\varphi\,dz.
$$

Combining the powers of $r$ gives

$$
I
=
\rho
\int_0^H dz
\int_0^{2\pi}d\varphi
\int_0^R r^3\,dr.
$$

The radial integral is

$$
\int_0^R r^3\,dr=\frac{R^4}{4}.
$$

Therefore,

$$
I
=
\rho H(2\pi)\frac{R^4}{4}
=
\frac{1}{2}\rho\pi R^4H.
$$

But the mass of the cylinder is

$$
M=\rho V=\rho\pi R^2H.
$$

Hence,

$$
\boxed{I=\frac{1}{2}MR^2}.
$$

Notice that the height $H$ disappears when the result is expressed in terms of the total mass $M$.


```python

# Visualization of the geometry for the moment-of-inertia integral

R, H = 2.0, 3.0

phi = np.linspace(0, 2*np.pi, 100)
z = np.linspace(0, H, 50)
Phi, Z = np.meshgrid(phi, z)

X = R*np.cos(Phi)
Y = R*np.sin(Phi)

fig = plt.figure(figsize=(7, 6))
ax = fig.add_subplot(111, projection='3d')

ax.plot_surface(X, Y, Z, alpha=0.18)

# axis of rotation
ax.plot([0, 0], [0, 0], [-0.4, H+0.4], linewidth=2)

# indicate a radial distance r
z0 = H/2
r0 = 1.4
phi0 = 0.4
x0 = r0*np.cos(phi0)
y0 = r0*np.sin(phi0)

ax.plot([0, x0], [0, y0], [z0, z0], linewidth=2)
ax.scatter([x0], [y0], [z0], s=35)
ax.text(x0/2, y0/2, z0+0.12, '$r$', fontsize=13)
ax.text(0, 0, H+0.55, 'axis of rotation', fontsize=11)

ax.set_xlabel('$x$')
ax.set_ylabel('$y$')
ax.set_zlabel('$z$')
ax.set_title('Cylinder rotating about its symmetry axis')
ax.set_box_aspect((1, 1, 1.2))
plt.show()
```

## Moment of inertia of a solid sphere

We can apply the same idea to a homogeneous solid sphere of radius $R$ and mass $M$, rotating about the $z$-axis.

Again,

$$
dI = r_\perp^2\,dm,
$$

where $r_\perp$ is the perpendicular distance of the mass element from the axis of rotation.

In spherical coordinates,

$$
r_\perp = r\sin\theta.
$$

For a homogeneous sphere,

$$
dm=\rho\,dV,
$$

and the spherical volume element is

$$
dV=r^2\sin\theta\,dr\,d\theta\,d\varphi.
$$

Therefore,

$$
dI
=
(r\sin\theta)^2
\rho r^2\sin\theta\,dr\,d\theta\,d\varphi.
$$

Thus,

$$
dI
=
\rho r^4\sin^3\theta\,dr\,d\theta\,d\varphi.
$$



### Setting up the integral

For the complete sphere,

$$
0\le r\le R,
$$

$$
0\le\theta\le\pi,
$$

and

$$
0\le\varphi\le2\pi.
$$

Hence,

$$
I
=
\rho
\int_0^R r^4\,dr
\int_0^\pi \sin^3\theta\,d\theta
\int_0^{2\pi}d\varphi.
$$

The three integrals are

$$
\int_0^R r^4\,dr=\frac{R^5}{5},
$$

$$
\int_0^\pi\sin^3\theta\,d\theta=\frac{4}{3},
$$

and

$$
\int_0^{2\pi}d\varphi=2\pi.
$$

Therefore,

$$
I
=
\rho\frac{R^5}{5}\frac{4}{3}(2\pi)
=
\frac{8}{15}\pi\rho R^5.
$$

The mass of the sphere is

$$
M=\rho\frac{4}{3}\pi R^3.
$$

Substituting this into the expression for $I$ gives

$$
\boxed{I=\frac{2}{5}MR^2}.
$$



### Why does $\sin^3\theta$ appear?

This is a useful example because two different geometrical factors occur.

The distance from the rotation axis gives

$$
r_\perp^2=(r\sin\theta)^2,
$$

while the spherical volume element contributes another factor

$$
dV=r^2\sin\theta\,dr\,d\theta\,d\varphi.
$$

Together,

$$
r_\perp^2\,dV
=
r^4\sin^3\theta\,dr\,d\theta\,d\varphi.
$$

This illustrates why choosing the correct distance from the **axis of rotation**, rather than simply the distance $r$ from the origin, is essential.
