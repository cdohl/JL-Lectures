# Complex Numbers


Complex numbers often look unfamiliar at first, but geometrically they are very simple:

> **A complex number is a point — or a vector — in a two-dimensional plane.**

They are especially useful for oscillations and waves. In physics, a harmonic oscillation can be written compactly using a complex exponential. At the end, we simply take the **real part** to obtain the physical displacement.

The aim of this notebook is therefore not abstract complex analysis. We only need enough complex-number mathematics to make harmonic oscillators easier to work with.

# 1. Why introduce a new number?

The equation

$$
x^2=-1
$$

has no solution among the real numbers.

We therefore define the imaginary unit

$$
\boxed{i^2=-1}.
$$

A complex number has the form

$$
\boxed{z=a+ib}
$$

where

- $a$ is the **real part**: $\operatorname{Re}(z)=a$,
- $b$ is the **imaginary part**: $\operatorname{Im}(z)=b$.

For example,

$$
z=3+2i
$$

has

$$
\operatorname{Re}(z)=3,
\qquad
\operatorname{Im}(z)=2.
$$

Despite the name *imaginary*, there is nothing mysterious about the geometry: $3+2i$ is simply represented by the point $(3,2)$ in the complex plane.

# 2. The Complex Plane

The horizontal axis represents the real part and the vertical axis the imaginary part.

For

$$
z=a+ib,
$$

the distance from the origin is the **magnitude**

$$
\boxed{|z|=\sqrt{a^2+b^2}}
$$

and the direction is described by the angle $\varphi$.

Thus a complex number contains the same geometrical information as a two-dimensional vector.

Use the sliders below to change the real and imaginary parts.

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
import ipywidgets as widgets
from IPython.display import display

def complex_number_plot(a=3.0, b=2.0):

    z = complex(a, b)
    r = abs(z)
    phi = np.angle(z)
    phi_deg = np.degrees(phi)

    fig, ax = plt.subplots(figsize=(6.5, 6.5))

    lim = 5.5
    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)

    # Complex number as an arrow
    ax.arrow(
        0, 0, a, b,
        length_includes_head=True,
        head_width=0.18,
        head_length=0.25,
        linewidth=2
    )

    # Components
    ax.plot([0, a], [b, b], 'k--', alpha=0.6)
    ax.plot([a, a], [0, b], 'k--', alpha=0.6)

    ax.text(a/2, -0.35 if b >= 0 else 0.25,
            rf'$\mathrm{{Re}}(z)={a:.1f}$',
            ha='center')

    ax.text(0.18 if a >= 0 else -0.18, b/2,
            rf'$\mathrm{{Im}}(z)={b:.1f}$',
            rotation=90, va='center',
            ha='left' if a >= 0 else 'right')

    # Angle arc
    if r > 0.05:
        radius = min(1.2, 0.35*r + 0.35)

        if phi_deg >= 0:
            theta1, theta2 = 0, phi_deg
        else:
            theta1, theta2 = phi_deg, 0

        arc = patches.Arc(
            (0,0),
            2*radius, 2*radius,
            angle=0,
            theta1=theta1,
            theta2=theta2,
            linewidth=2
        )
        ax.add_patch(arc)

        mid = phi/2
        ax.text(
            1.25*radius*np.cos(mid),
            1.25*radius*np.sin(mid),
            r'$\varphi$',
            ha='center', va='center'
        )

    ax.text(
        0.03, 0.96,
        rf'$z={a:.1f}{b:+.1f}i$' '\n'
        rf'$|z|={r:.2f}$' '\n'
        rf'$\varphi={phi_deg:.1f}^\circ$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9)
    )

    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect('equal')
    ax.set_xlabel('real part')
    ax.set_ylabel('imaginary part')
    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)

a_slider = widgets.FloatSlider(
    value=3, min=-5, max=5, step=0.5,
    description='Re(z):'
)

b_slider = widgets.FloatSlider(
    value=2, min=-5, max=5, step=0.5,
    description='Im(z):'
)

out = widgets.interactive_output(
    complex_number_plot,
    {'a': a_slider, 'b': b_slider}
)

display(widgets.HBox([a_slider, b_slider]), out)
```

# 3. Cartesian and Polar Form

The same complex number can be described in two ways.

### Cartesian form

$$
\boxed{z=a+ib}
$$

### Polar form

From the geometry,

$$
a=r\cos\varphi,
\qquad
b=r\sin\varphi.
$$

Therefore,

$$
z=r\cos\varphi+i\,r\sin\varphi
$$

or

$$
\boxed{
z=r(\cos\varphi+i\sin\varphi).
}
$$

Here,

$$
r=|z|
$$

is the magnitude and $\varphi$ is the angle.

For example, for

$$
z=1+i
$$

we obtain

$$
|z|=\sqrt{1^2+1^2}=\sqrt2
$$

and

$$
\varphi=45^\circ=\frac{\pi}{4}.
$$

# 4. Adding Complex Numbers

Addition is particularly simple:

$$
(a+ib)+(c+id)
=
(a+c)+i(b+d).
$$

We simply add the real parts and the imaginary parts separately.

For example,

$$
(2+i)+(1+2i)=3+3i.
$$

Geometrically, this is exactly the same as adding two vectors.

Use the sliders below to add two complex numbers.

```python
def complex_addition(a=2.0, b=1.0, c=1.0, d=2.0):

    z1 = complex(a,b)
    z2 = complex(c,d)
    z3 = z1 + z2

    fig, ax = plt.subplots(figsize=(6.5,6.5))

    def arrow(x0, y0, dx, dy, label):
        ax.arrow(
            x0, y0, dx, dy,
            length_includes_head=True,
            head_width=0.16,
            head_length=0.22,
            linewidth=2
        )
        ax.text(
            x0 + 0.52*dx,
            y0 + 0.52*dy + 0.18,
            label,
            fontsize=15
        )

    # z1 and z2 from origin
    arrow(0, 0, a, b, r'$z_1$')
    arrow(0, 0, c, d, r'$z_2$')

    # z1 + z2
    arrow(0, 0, z3.real, z3.imag, r'$z_1+z_2$')

    # Parallelogram construction
    ax.plot([a, z3.real], [b, z3.imag], 'k--', alpha=0.5)
    ax.plot([c, z3.real], [d, z3.imag], 'k--', alpha=0.5)

    ax.axhline(0, color='k', linewidth=1)
    ax.axvline(0, color='k', linewidth=1)

    ax.text(
        0.03, 0.96,
        rf'$z_1={a:.1f}{b:+.1f}i$' '\n'
        rf'$z_2={c:.1f}{d:+.1f}i$' '\n'
        rf'$z_1+z_2={z3.real:.1f}{z3.imag:+.1f}i$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9)
    )

    ax.set_xlim(-6,6)
    ax.set_ylim(-6,6)
    ax.set_aspect('equal')
    ax.set_xlabel('real part')
    ax.set_ylabel('imaginary part')
    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)

a1 = widgets.FloatSlider(value=2, min=-3, max=3, step=0.5, description='Re(z₁):')
b1 = widgets.FloatSlider(value=1, min=-3, max=3, step=0.5, description='Im(z₁):')
a2 = widgets.FloatSlider(value=1, min=-3, max=3, step=0.5, description='Re(z₂):')
b2 = widgets.FloatSlider(value=2, min=-3, max=3, step=0.5, description='Im(z₂):')

out2 = widgets.interactive_output(
    complex_addition,
    {'a':a1, 'b':b1, 'c':a2, 'd':b2}
)

display(
    widgets.VBox([
        widgets.HBox([a1,b1]),
        widgets.HBox([a2,b2])
    ]),
    out2
)
```

# 5. Multiplication by $i$ Means Rotation

Consider

$$
z=a+ib.
$$

Multiplying by $i$ gives

$$
iz=ia+i^2b=-b+ia.
$$

Thus

$$
(a,b)\longrightarrow(-b,a).
$$

This is a rotation by $90^\circ$ in the complex plane.

For example,

$$
1 \xrightarrow{\times i} i
\xrightarrow{\times i} -1
\xrightarrow{\times i} -i
\xrightarrow{\times i} 1.
$$

This geometrical interpretation of multiplication is one reason complex numbers are so useful for oscillations and rotations.

# 6. Euler's Formula

The key relation connecting complex numbers with trigonometry is

$$
\boxed{
e^{i\varphi}
=
\cos\varphi+i\sin\varphi.
}
$$

This is **Euler's formula**.

It says that the complex number $e^{i\varphi}$ lies on the unit circle:

$$
|e^{i\varphi}|=1.
$$

Its real and imaginary parts are

$$
\boxed{
\operatorname{Re}(e^{i\varphi})=\cos\varphi
}
$$

and

$$
\boxed{
\operatorname{Im}(e^{i\varphi})=\sin\varphi.
}
$$

Therefore a complex exponential is simply a convenient way of describing motion around a circle.

```python
def euler_plot(phi=45):

    p = np.deg2rad(phi)
    x = np.cos(p)
    y = np.sin(p)

    fig, ax = plt.subplots(figsize=(6.5,6.5))

    theta = np.linspace(0,2*np.pi,400)
    ax.plot(np.cos(theta), np.sin(theta), linewidth=1.5)

    ax.arrow(
        0,0,x,y,
        length_includes_head=True,
        head_width=0.06,
        head_length=0.09,
        linewidth=2
    )

    ax.plot([x,x],[0,y],'k--',alpha=0.6)
    ax.plot([0,x],[y,y],'k--',alpha=0.6)

    ax.axhline(0,color='k',linewidth=1)
    ax.axvline(0,color='k',linewidth=1)

    ax.text(x/2, -0.12, r'$\cos\varphi$', ha='center')
    ax.text(0.08, y/2, r'$\sin\varphi$', va='center')

    arc = patches.Arc(
        (0,0), 0.7,0.7,
        angle=0,
        theta1=0,
        theta2=phi,
        linewidth=2
    )
    ax.add_patch(arc)

    ax.text(
        0.03,0.96,
        rf'$e^{{i\varphi}}={x:.3f}{y:+.3f}i$' '\n'
        rf'$\cos\varphi={x:.3f}$' '\n'
        rf'$\sin\varphi={y:.3f}$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9)
    )

    ax.set_xlim(-1.35,1.35)
    ax.set_ylim(-1.35,1.35)
    ax.set_aspect('equal')
    ax.set_xlabel('real part')
    ax.set_ylabel('imaginary part')
    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)

phi_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description=r'$\varphi$:'
)

out3 = widgets.interactive_output(
    euler_plot,
    {'phi':phi_slider}
)

display(phi_slider,out3)
```

# 7. Why Complex Numbers Make Trigonometry Easier

Suppose we want to combine two angles $\alpha$ and $\beta$.

With ordinary trigonometry we need to remember identities such as

$$
\cos(\alpha+\beta)
=
\cos\alpha\cos\beta-\sin\alpha\sin\beta
$$

and

$$
\sin(\alpha+\beta)
=
\sin\alpha\cos\beta+\cos\alpha\sin\beta.
$$

With complex numbers, angle addition follows from the ordinary rule for exponentials:

$$
e^{i\alpha}e^{i\beta}
=
e^{i(\alpha+\beta)}.
$$

Using Euler's formula,

$$
(\cos\alpha+i\sin\alpha)
(\cos\beta+i\sin\beta)
=
\cos(\alpha+\beta)+i\sin(\alpha+\beta).
$$

Multiplying the left side gives

$$
\cos\alpha\cos\beta
-\sin\alpha\sin\beta
+i\left(
\sin\alpha\cos\beta
+\cos\alpha\sin\beta
\right).
$$

The **real part** immediately gives the cosine addition rule, and the **imaginary part** gives the sine addition rule.

So instead of memorizing several trigonometric rules, we can often use the much simpler algebra of exponentials.

# 8. Harmonic Oscillations

A real harmonic oscillation is usually written as

$$
x(t)=A\cos(\omega t+\varphi).
$$

Using Euler's formula we can instead write a complex oscillation

$$
\boxed{
z(t)=A e^{i(\omega t+\varphi)}.
}
$$

The physical displacement is its real part:

$$
\boxed{
x(t)=\operatorname{Re}[z(t)].
}
$$

This may initially look more complicated, but differentiation becomes extremely simple:

$$
\frac{d}{dt}e^{i\omega t}
=
i\omega e^{i\omega t}.
$$

Therefore, if

$$
z(t)=Ae^{i\omega t},
$$

then

$$
\dot z(t)=i\omega Ae^{i\omega t}
$$

and

$$
\ddot z(t)
=
-\omega^2Ae^{i\omega t}.
$$

Hence

$$
\boxed{
\ddot z=-\omega^2z.
}
$$

This is exactly the equation of a harmonic oscillator.

The complicated-looking sine and cosine differentiation has been reduced to multiplication by $i\omega$.

# 9. The Complex Oscillation and Its Real Part

The complex number

$$
z(t)=Ae^{i\omega t}
$$

rotates around a circle in the complex plane.

Its projection onto the real axis is

$$
x(t)=A\cos(\omega t).
$$

Thus a harmonic oscillation can be viewed as the **projection of uniform circular motion**.

```python
def harmonic_plot(phi=45):

    p = np.deg2rad(phi)
    x = np.cos(p)
    y = np.sin(p)

    fig, ax = plt.subplots(figsize=(7,5.5))

    theta = np.linspace(0,2*np.pi,400)
    ax.plot(np.cos(theta),np.sin(theta),linewidth=1.5)

    ax.arrow(
        0,0,x,y,
        length_includes_head=True,
        head_width=0.06,
        head_length=0.09,
        linewidth=2
    )

    # Projection onto real axis
    ax.plot([x,x],[0,y],'k--',linewidth=1.5)
    ax.plot(x,0,'o',markersize=7)

    ax.axhline(0,color='k',linewidth=1)
    ax.axvline(0,color='k',linewidth=1)

    ax.text(
        x, -0.16,
        r'$x=A\cos(\omega t+\varphi_0)$',
        ha='center'
    )

    ax.text(
        0.03,0.95,
        r'$z=Ae^{i(\omega t+\varphi_0)}$' '\n'
        rf'$\mathrm{{Re}}(z)/A={x:.3f}$',
        transform=ax.transAxes,
        va='top',
        bbox=dict(boxstyle='round',facecolor='white',alpha=0.9)
    )

    ax.set_xlim(-1.4,1.4)
    ax.set_ylim(-1.4,1.4)
    ax.set_aspect('equal')
    ax.set_xlabel('real part')
    ax.set_ylabel('imaginary part')
    ax.grid(alpha=0.2)

    plt.show()
    plt.close(fig)

phase_slider = widgets.FloatSlider(
    value=45,
    min=0,
    max=360,
    step=5,
    description='phase:'
)

out4 = widgets.interactive_output(
    harmonic_plot,
    {'phi':phase_slider}
)

display(phase_slider,out4)
```

# 10. A Very Useful Rule for Oscillations

For a complex exponential,

$$
z(t)=Ae^{i\omega t},
$$

differentiation with respect to time is equivalent to multiplication by $i\omega$:

$$
\boxed{
\frac{d}{dt}
\quad\longleftrightarrow\quad
i\omega.
}
$$

Differentiating twice gives

$$
\boxed{
\frac{d^2}{dt^2}
\quad\longleftrightarrow\quad
-\omega^2.
}
$$

This simple rule will become extremely useful in the physics lecture when harmonic oscillators and driven oscillations are discussed.
