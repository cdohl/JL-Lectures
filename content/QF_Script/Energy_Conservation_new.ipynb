## 4.1 Work: force acting through a displacement

For a small displacement $d\vec r$, a force $\vec F$ does the work

$$dW=\vec F\cdot d\vec r=|\vec F|\,|d\vec r|\cos\theta.$$

Over a path $C$ from position 1 to position 2,

$$\boxed{W_{1\to2}=\int_C\vec F\cdot d\vec r}.$$

For a **constant force** acting over a straight displacement $s$:

$$\boxed{W=Fs\cos\theta}.$$

- $\theta=0^\circ$: **positive** work (force acts along the motion).
- $\theta=90^\circ$: **zero** work (force is perpendicular to the motion).
- $\theta=180^\circ$: **negative** work (force opposes the motion).

**Important:** Work is a scalar and can be positive, negative, or zero. The unit is $\mathrm{N\,m}=\mathrm{J}$.

### Interactive 1 — The angle determines the work

Change the force magnitude, displacement, and angle. The arrow shows the force relative to a displacement to the right. **Predict the sign of the work before moving the angle slider.**

```python
import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

plt.rcParams.update({"figure.figsize": (8, 4), "axes.grid": True})

def plot_work(force=4.0, distance=2.0, angle=60):
    theta = np.deg2rad(angle)
    result = force * distance * np.cos(theta)
    fig, ax = plt.subplots(figsize=(8, 3.4))
    ax.arrow(0, 0, 2, 0, width=0.02, head_width=0.14,
             length_includes_head=True, color='tab:blue', label='Displacement')
    scale = 1.5 * force / 10
    ax.arrow(0, 0, scale*np.cos(theta), scale*np.sin(theta),
             width=0.02, head_width=0.14, length_includes_head=True,
             color='tab:orange', label='Force')
    ax.plot([], [], color='tab:blue', label='Displacement')
    ax.plot([], [], color='tab:orange', label='Force')
    ax.set(xlim=(-2, 2.5), ylim=(-2, 2), aspect='equal', xlabel='Diagram (arbitrary length)',
           title=f'W = {force:.1f} N × {distance:.1f} m × cos({angle}°) = {result:.3f} J')
    ax.legend(loc='lower right'); plt.show()

widgets.interact(plot_work,
    force=widgets.FloatSlider(value=4, min=0, max=10, step=0.5, description='Force (N)'),
    distance=widgets.FloatSlider(value=2, min=0, max=5, step=0.25, description='Path (m)'),
    angle=widgets.IntSlider(value=60, min=0, max=180, step=5, description='Angle (°)'));
```

### Worked example 1 — Pulling against friction

A body moves $3\,\mathrm{m}$ horizontally. An applied force of $8\,\mathrm{N}$ makes an angle of $60^\circ$ with the direction of motion. The body travels at constant speed.

The applied force does work

$$W_\mathrm{applied}=Fs\cos60^\circ =8\times3\times0.5=12\,\mathrm{J}.$$

For constant speed, the net work is zero (no change in kinetic energy), so friction does

$$W_\mathrm{friction}=-12\,\mathrm{J}.$$

The applied work is transferred to other forms of energy (largely internal/thermal energy); it does **not** appear as an increase in kinetic energy.

**Class question:** What happens to the work if the applied force becomes perpendicular to the path?

## 4.2 Work and kinetic energy

Newton's second law (constant mass) gives

$$\vec F_\mathrm{net}=m\frac{d\vec v}{dt}.$$

Use $d\vec r=\vec v\,dt$ to rewrite the net work:

$$W_\mathrm{net}=\int \vec F_\mathrm{net}\cdot d\vec r
=\int m\frac{d\vec v}{dt}\cdot\vec v\,dt
=\int m\vec v\cdot d\vec v.$$

Since $d(v^2)=2\vec v\cdot d\vec v$,

$$\boxed{W_\mathrm{net}=\tfrac12 mv_2^2-\tfrac12 mv_1^2=\Delta E_\mathrm{kin}}.$$

We define **kinetic energy** as

$$\boxed{E_\mathrm{kin}=\tfrac12 mv^2}.$$

Notice that **net work**, not the work of just one selected force, changes kinetic energy.

### Worked example 2 — Accelerating an object

A $0.50\,\mathrm{kg}$ object increases its speed from $1$ to $3\,\mathrm{m/s}$.

$$\Delta E_\mathrm{kin}=\frac{0.50}{2}(3^2-1^2)=2.0\,\mathrm{J}.$$

The net force must therefore perform $2.0\,\mathrm{J}$ of work, regardless of the detailed path taken between these speeds.

## 4.3 Potential energy and conservative forces

A force is **conservative** if its work depends only on the initial and final positions. For such a force we can define potential energy by

$$\boxed{E_\mathrm{pot}(\vec r_2)-E_\mathrm{pot}(\vec r_1)
=-\int_{\vec r_1}^{\vec r_2}\vec F_\mathrm{cons}\cdot d\vec r}.$$

Examples:

**Gravity near Earth's surface** (constant $g$):

$$\boxed{E_\mathrm{grav}=mgh}\quad\text{(taking }E_\mathrm{grav}(0)=0\text{)}.$$

**Ideal spring** (Hooke's law $F_x=-kx$):

$$\boxed{E_\mathrm{spring}=\tfrac12 kx^2}\quad\text{(taking }E_\mathrm{spring}(0)=0\text{)}.$$

The zero of potential energy is arbitrary; **differences** in potential energy are what matter. Friction is not conservative and cannot generally be represented by a single position-dependent potential energy.

## 4.4 Conservation of mechanical energy

Define the mechanical energy

$$E_\mathrm{mech}=E_\mathrm{kin}+E_\mathrm{pot}.$$

If **only conservative forces do work**, then

$$\boxed{E_\mathrm{kin}+E_\mathrm{pot}=\mathrm{constant}}.$$

If nonconservative forces (e.g., friction or muscular driving forces) do work, then

$$\boxed{\Delta E_\mathrm{mech}=W_\mathrm{noncons}}.$$

The **total energy** of an isolated system remains conserved; mechanical energy can be converted into thermal, chemical, or other forms.

### Interactive 2 — Falling without air resistance

Move the object to different heights. The sum of kinetic and gravitational potential energy stays constant. The example assumes release from rest at height $h_0$.

```python
def falling_energy(h0=5.0, fraction=0.6):
    m, g = 1.0, 9.81
    h = fraction * h0
    ep = m*g*h
    ek = m*g*(h0-h)
    speed = np.sqrt(2*g*(h0-h))
    fig, axes = plt.subplots(1, 2, figsize=(9, 3.5))
    z = np.linspace(0, h0, 250)
    axes[0].plot(z, m*g*z, label='Potential energy')
    axes[0].plot(z, m*g*(h0-z), label='Kinetic energy')
    axes[0].plot(z, np.full_like(z, m*g*h0), '--', label='Total mechanical')
    axes[0].axvline(h, color='0.45', linestyle=':')
    axes[0].set(xlabel='Height (m)', ylabel='Energy (J)', title='Energy versus height')
    axes[0].legend(fontsize=8)
    axes[1].bar(['Potential', 'Kinetic', 'Total'], [ep, ek, ep+ek],
                color=['tab:green', 'tab:orange', 'tab:blue'])
    axes[1].set(ylabel='Energy (J)', ylim=(0, max(1,m*g*h0)*1.15),
                title=f'At h = {h:.2f} m: speed = {speed:.2f} m/s')
    plt.tight_layout(); plt.show()

widgets.interact(falling_energy,
    h0=widgets.FloatSlider(value=5, min=1, max=10, step=0.5, description='Start h (m)'),
    fraction=widgets.FloatSlider(value=0.6, min=0, max=1, step=0.05, description='h / start'));
```

### Worked example 3 — A falling seed (idealized)

A $0.01\,\mathrm{kg}$ seed falls from rest through $2\,\mathrm{m}$ **without drag**. How fast does it move after falling that distance?

$$m g\Delta h=\tfrac12 mv^2\quad\Rightarrow\quad
v=\sqrt{2g\Delta h}=\sqrt{2\times9.81\times2}\approx6.26\,\mathrm{m/s}.$$

The mass cancels. **Reality check:** air drag can be important for small seeds; then this idealized speed overestimates the actual value.

## 4.5 Power: how fast is work done?

Instantaneous power is the rate at which a force does work:

$$\boxed{P=\frac{dW}{dt}=\vec F\cdot\vec v}.$$

Average power over a time interval $\Delta t$ is

$$\boxed{\overline P=\frac{\Delta W}{\Delta t}}.$$

**Units:** $1\,\mathrm{W}=1\,\mathrm{J/s}$. Work or energy is measured in joules; power in watts. A large power over a short time may transfer the same energy as a small power over a long time.

### Interactive 3 — Instantaneous versus average power

Suppose a time-dependent force $F(t)=F_0[1+\alpha\sin(2\pi t/T)]$ acts in the direction of a constant speed $v$. Then $P(t)=F(t)v$.

```python
def power_demo(F0=3.0, speed=1.0, alpha=0.5):
    T = 4.0
    t = np.linspace(0, T, 400)
    p = F0 * (1 + alpha*np.sin(2*np.pi*t/T)) * speed
    avg = np.trapezoid(p, t) / T if hasattr(np, 'trapezoid') else np.trapz(p, t) / T
    work = avg * T
    fig, ax = plt.subplots(figsize=(8, 3.5))
    ax.plot(t, p, label='Instantaneous power P(t)')
    ax.axhline(avg, color='tab:orange', linestyle='--', label='Average power')
    ax.set(xlabel='Time (s)', ylabel='Power (W)',
           title=f'Work in {T:g} s = {work:.2f} J; average power = {avg:.2f} W')
    ax.legend(); plt.show()

widgets.interact(power_demo,
    F0=widgets.FloatSlider(value=3, min=0.5, max=10, step=0.5, description='Mean F (N)'),
    speed=widgets.FloatSlider(value=1, min=0.1, max=3, step=0.1, description='Speed (m/s)'),
    alpha=widgets.FloatSlider(value=0.5, min=0, max=0.95, step=0.05, description='Variation'));
```

### Worked example 4 — Different powers, same energy

Lifting a $2\,\mathrm{kg}$ mass upward by $1\,\mathrm{m}$ at constant speed requires

$$W=mgh=2\times9.81\times1=19.62\,\mathrm{J}.$$

- If the lift takes $2\,\mathrm{s}$, average power is $9.81\,\mathrm{W}$.
- If it takes $10\,\mathrm{s}$, average power is $1.962\,\mathrm{W}$.

The work done is the same; the power differs by a factor of five.

## 4.6 Biology connection — Work done by a molecular motor

Cells contain molecular motors that convert chemical free energy into mechanical work. Consider an **illustrative** motor that exerts a force of $5\,\mathrm{pN}$ along a displacement of $8\,\mathrm{nm}$.

First convert units:

$$5\,\mathrm{pN}=5\times10^{-12}\,\mathrm{N},\qquad
8\,\mathrm{nm}=8\times10^{-9}\,\mathrm{m}.$$

The mechanical work per step is

$$\boxed{W=Fs=(5\times10^{-12})(8\times10^{-9})=4\times10^{-20}\,\mathrm{J}}.$$

If the motor makes $100$ such steps each second under the same load, the average mechanical output power is

$$\boxed{\overline P=100W=4\times10^{-18}\,\mathrm{W}}.$$

These are *example parameters*, not a universal description of every molecular motor. Actual motor force, displacement, stepping rate and efficiency depend on the motor and conditions.

### Interactive 4 — Force, step size, and stepping rate

```python
def motor_demo(force_pN=5.0, step_nm=8.0, rate=100):
    work_J = force_pN * 1e-12 * step_nm * 1e-9
    power_W = work_J * rate
    print(f'Work per step:     {work_J:.3e} J')
    print(f'Work per step:     {work_J/1e-21:.1f} zJ (zeptojoules)')
    print(f'Mean output power: {power_W:.3e} W')
    print(f'Mean output power: {power_W/1e-18:.2f} aW (attowatts)')

widgets.interact(motor_demo,
    force_pN=widgets.FloatSlider(value=5, min=0, max=10, step=0.5, description='Force (pN)'),
    step_nm=widgets.FloatSlider(value=8, min=1, max=20, step=1, description='Step (nm)'),
    rate=widgets.IntSlider(value=100, min=0, max=300, step=10, description='Steps/s'));
```

### Biological interpretation

A molecular motor's **mechanical output energy** is not the same quantity as its **chemical input free energy**. For a motor operating cyclically,

$$\eta=\frac{\text{useful mechanical output work}}{\text{chemical free-energy input}}$$

is a possible definition of efficiency, provided the energy accounting and conditions are specified. The difference appears in other energy transfers and dissipation. An external force opposing the motion makes **negative** work on the moving motor; the motor may do positive work against that load.

**Class question:** If a motor does the same work per step but steps twice as often, what happens to its average output power?

## 4.7 Equilibrium from potential energy

In one dimension, a conservative force is related to potential energy by

$$\boxed{F_x(x)=-\frac{dE_\mathrm{pot}}{dx}}.$$

At an equilibrium position $x_*$,

$$\boxed{\left.\frac{dE_\mathrm{pot}}{dx}\right|_{x_*}=0}.$$

- **Stable equilibrium:** a local potential-energy minimum ($E''_\mathrm{pot}(x_*)>0$ for a nondegenerate equilibrium).
- **Unstable equilibrium:** a local maximum ($E''_\mathrm{pot}(x_*)<0$).
- If the second derivative is zero, a higher-order analysis may be needed.

### Interactive 5 — Stable versus unstable

The potential $E(x)=\frac14 x^4-\frac12 b x^2$ illustrates how the shape and equilibrium positions change with the parameter $b$.

### Stable versus unstable equilibrium — geometric picture

At the bottom of a potential well, small displacements increase potential energy. At the top of a potential hill, small displacements decrease potential energy. Both positions satisfy $dE_{\mathrm{pot}}/dx=0$, but only the minimum is stable.

**Stable equilibrium**



*A ball near a local minimum of potential energy: small displacements generate a restoring force.*

**Unstable equilibrium**



*A ball near a local maximum of potential energy: small displacements lead away from equilibrium.*

```python
def equilibrium_demo(b=1.0):
    x = np.linspace(-2, 2, 500)
    E = x**4 / 4 - b*x**2 / 2
    F = -x**3 + b*x
    fig, axes = plt.subplots(1, 2, figsize=(9, 3.3))
    axes[0].plot(x, E); axes[0].axhline(0, linewidth=0.7, color='0.5')
    axes[0].set(xlabel='Position x', ylabel='Potential energy (arb. units)', title='Energy landscape')
    axes[1].plot(x, F); axes[1].axhline(0, linewidth=0.7, color='0.5')
    axes[1].set(xlabel='Position x', ylabel='Force (arb. units)', title='F(x) = −dE/dx')
    if b > 0:
        print(f'Equilibria: x = 0 (unstable), x = ±{np.sqrt(b):.3f} (stable).')
    elif b < 0:
        print('Only x = 0 is an equilibrium (stable).')
    else:
        print('Only x = 0 is an equilibrium (stable, but E″(0) = 0).')
    plt.tight_layout(); plt.show()

widgets.interact(equilibrium_demo,
    b=widgets.FloatSlider(value=1, min=-1, max=2, step=0.1, description='Parameter b'));
```
