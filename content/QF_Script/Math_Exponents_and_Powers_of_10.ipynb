# Exponents, Powers of Ten and Logarithms

In physics, we often encounter very large and very small numbers. Before learning how to write and calculate with these numbers efficiently, we will review the basic rules for **powers and exponents**.

A power is written as

$$
a^n
$$

where $a$ is the **base** and $n$ is the **exponent**.

For a positive integer exponent,

$$
a^n=\underbrace{a\cdot a\cdot a\cdots a}_{n\text{ times}}.
$$

For example,

$$
2^4=2\cdot2\cdot2\cdot2=16.
$$

## 1. Multiplying powers

If two powers have the **same base**, their exponents are added:

$$
a^m\cdot a^n=a^{m+n}.
$$

### Example

$$
2^3\cdot2^4=2^{3+4}=2^7=128.
$$

This works because

$$
(2\cdot2\cdot2)(2\cdot2\cdot2\cdot2)=2^7.
$$

## 2. Dividing powers

When powers with the same base are divided, their exponents are subtracted:

$$
\frac{a^m}{a^n}=a^{m-n}.
$$

### Example

$$
\frac{3^5}{3^2}=3^{5-2}=3^3=27.
$$

This rule also explains negative exponents:

$$
a^{-n}=\frac{1}{a^n}.
$$

For example,

$$
2^{-3}=\frac{1}{2^3}=\frac18.
$$

## 3. Raising a power to a power

When a power is raised to another power, the exponents are multiplied:

$$
\left(a^m\right)^n=a^{mn}.
$$

### Example

$$
(2^3)^4=2^{3\cdot4}=2^{12}.
$$

## 4. Adding and subtracting powers

There is **no general rule for adding the exponents** when powers are added.

For example,

$$
2^3+2^4\neq2^7.
$$

Instead, calculate the powers separately:

$$
2^3+2^4=8+16=24.
$$

However, if the powers are identical, they can be collected like any other common factor:

$$
3\cdot10^4+5\cdot10^4=(3+5)\cdot10^4=8\cdot10^4.
$$

This will become particularly useful when we work with powers of 10.

## Summary of the exponent rules

| Operation | Rule |
|---|---|
| Multiplication | $a^m a^n=a^{m+n}$ |
| Division | $\displaystyle\frac{a^m}{a^n}=a^{m-n}$ |
| Power of a power | $(a^m)^n=a^{mn}$ |
| Negative exponent | $\displaystyle a^{-n}=\frac{1}{a^n}$ |
| Zero exponent | $a^0=1$ |

### Be careful!

In general,

$$
a^m+a^n\neq a^{m+n}.
$$

# Powers of 10

Powers of 10 are particularly useful in physics.

Consider

$$
10^3=1000
$$

and

$$
10^{-3}=\frac{1}{1000}=0.001.
$$

Each increase of the exponent by 1 multiplies the number by 10:

$$
10^0=1,\qquad
10^1=10,\qquad
10^2=100,\qquad
10^3=1000.
$$

Likewise,

$$
10^{-1}=0.1,\qquad
10^{-2}=0.01,\qquad
10^{-3}=0.001.
$$

The exponent therefore tells us the **order of magnitude** of a number.

## Why are powers of 10 useful in physics?

Physical quantities can differ enormously in size. For example, approximately:

| Quantity | Size |
|---|---:|
| Diameter of a human hair | $7\times10^{-5}\,\mathrm{m}$ |
| Diameter of a red blood cell | $8\times10^{-6}\,\mathrm{m}$ |
| Diameter of a hydrogen atom | $1\times10^{-10}\,\mathrm{m}$ |
| Distance travelled by light in one second | $3\times10^8\,\mathrm{m}$ |

Writing these numbers using powers of 10 makes their **scale** immediately visible.

For example,

$$
0.000008\,\mathrm{m}=8\times10^{-6}\,\mathrm{m}.
$$

Instead of counting zeros, we can calculate directly with the exponent.

# Logarithms — undoing exponentiation

So far we have used powers such as

$$
10^3=1000
$$

or

$$
2^5=32.
$$

A logarithm asks the **inverse question**:

> To what power must the base be raised to obtain this number?

For example,

$$
10^3=1000
$$

means exactly the same as

$$
\log_{10}(1000)=3.
$$

Similarly,

$$
2^5=32
$$

means

$$
\log_2(32)=5.
$$

In general,

$$
a^x=y
\qquad\Longleftrightarrow\qquad
\log_a(y)=x.
$$

## `log` and `ln`

Two logarithms occur particularly often.

### Base 10

In this notebook we write

$$
\log(x)=\log_{10}(x).
$$

It is especially convenient when working with **orders of magnitude**:

$$
\log(10)=1,\qquad
\log(100)=2,\qquad
\log(1000)=3.
$$

Also,

$$
\log(0.1)=-1
$$

because

$$
10^{-1}=0.1.
$$

### Natural logarithm

The **natural logarithm**

$$
\ln(x)
$$

uses the number

$$
e\approx2.718
$$

as its base:

$$
\ln(x)=\log_e(x).
$$

Therefore,

$$
\ln(e)=1,
\qquad
\ln(e^2)=2.
$$

The natural logarithm is particularly important for **exponential growth and decay**, which occur frequently in biology.

## Logarithms turn multiplication into addition

One of the most useful properties is

$$
\log(ab)=\log(a)+\log(b).
$$

Similarly,

$$
\log\left(\frac{a}{b}\right)
=
\log(a)-\log(b)
$$

and

$$
\log(a^n)=n\log(a).
$$

The same rules hold for $\ln$.

This last rule is particularly useful because it allows us to bring an exponent down in front of the logarithm.

# Solving exponential equations

Suppose a bacterial population doubles every generation:

$$
N=2^n.
$$

If we know $n$, calculating $N$ is easy.

But suppose we know the population and want to determine the number of generations:

$$
1000=2^n.
$$

The unknown quantity is now in the **exponent**.

Take the natural logarithm of both sides:

$$
\ln(1000)=\ln(2^n).
$$

Using

$$
\ln(a^n)=n\ln(a),
$$

we obtain

$$
\ln(1000)=n\ln(2),
$$

and therefore

$$
n=\frac{\ln(1000)}{\ln(2)}.
$$

The same calculation would work with `log`:

$$
n=\frac{\log(1000)}{\log(2)}.
$$

The answer is the same because the choice of logarithm base cancels in the ratio.

# Natural logarithms and exponential growth

Continuous growth or decay is often written as

$$
N(t)=N_0e^{kt}.
$$

Taking the natural logarithm gives

$$
\ln N
=
\ln N_0 + kt.
$$

This is important because an exponential relation has become a **linear relation**.

Compare

$$
\ln N=\ln N_0+kt
$$

with the equation of a straight line

$$
y=a+mx.
$$

We can identify:

- $y\leftrightarrow\ln N$
- $x\leftrightarrow t$
- intercept $a\leftrightarrow\ln N_0$
- slope $m\leftrightarrow k$

Thus a plot of $\ln N$ against $t$ is a straight line if the population grows exponentially.

```python
import numpy as np
import matplotlib.pyplot as plt

t = np.linspace(0, 5, 100)

N0 = 2
k = 0.7

N = N0 * np.exp(k*t)

plt.figure(figsize=(7, 4))
plt.plot(t, N)
plt.xlabel("time")
plt.ylabel("N")
plt.title("Exponential growth")
plt.grid()
plt.show()

plt.figure(figsize=(7, 4))
plt.plot(t, np.log(N))
plt.xlabel("time")
plt.ylabel("ln(N)")
plt.title("The same data after taking the natural logarithm")
plt.grid()
plt.show()
```

### What do you observe?

The first graph is curved because the population grows exponentially.

The second graph is a straight line because

$$
\ln N=\ln N_0+kt.
$$

Its slope is $k$.

This idea will appear repeatedly later in the course: logarithms can reveal simple relationships hidden inside rapidly varying data.

# Logarithms and power laws

Logarithms are also useful for relationships such as

$$
y=Cx^n.
$$

Taking the logarithm gives

$$
\ln y
=
\ln C+n\ln x.
$$

Again we obtain a straight-line equation.

If we plot $\ln y$ against $\ln x$, the slope is the exponent $n$.

For example, if

$$
y\propto x^2,
$$

the slope of a log-log plot is 2.

If

$$
y\propto\frac{1}{x^2}=x^{-2},
$$

the slope is $-2$.

# Summary

### Powers and logarithms are inverse operations

$$
a^x=y
\qquad\Longleftrightarrow\qquad
\log_a(y)=x.
$$

### Base-10 logarithm

$$
\log(x)=\log_{10}(x)
$$

is especially useful for powers of ten and orders of magnitude.

### Natural logarithm

$$
\ln(x)=\log_e(x)
$$

is especially useful for exponential growth and decay.

### Important rules

$$
\ln(ab)=\ln a+\ln b
$$

$$
\ln\left(\frac{a}{b}\right)=\ln a-\ln b
$$

$$
\ln(a^n)=n\ln a.
$$

These rules allow us to solve exponential equations and to transform exponential and power-law relationships into straight lines.
