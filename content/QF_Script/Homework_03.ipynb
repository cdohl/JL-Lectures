## 💻 10. Programming — obtain velocity and acceleration from a trajectory
Imagine that a camera records the position of a particle moving on a circle. The experiment gives us only a sequence of measured positions

$$
x(t),\qquad y(t).
$$

Can we recover the velocity and acceleration from these data?

The program below creates such position data and then uses numerical differentiation to estimate

$$
v_x=\frac{dx}{dt},\qquad v_y=\frac{dy}{dt}
$$

and

$$
a_x=\frac{dv_x}{dt},\qquad a_y=\frac{dv_y}{dt}.
$$

### Tasks

**a)** Run the program and inspect the three graphs.

**b)** Is the speed approximately constant even though $v_x$ and $v_y$ change?

**c)** Compare the numerical acceleration magnitude with the theoretical value

$$
a_c=R\omega^2.
$$

**d)** Change $\omega$ from $1$ to $2\,\mathrm{rad/s}$. By what factor does the acceleration change?

**e)** Change $R$ while keeping $\omega$ constant. How does the acceleration depend on $R$?

**f)** Look at the position and acceleration vectors at one instant. What relationship do you observe between them? Connect your observation with

$$
\vec a=-\omega^2\vec r.
$$

```python
import numpy as np
import matplotlib.pyplot as plt

# Parameters
R = 2.0          # radius in m
omega = 1.0      # angular velocity in rad/s

# Sampled times
t = np.linspace(0, 2*np.pi/omega, 200)

# "Measured" positions
x = R * np.cos(omega*t)
y = R * np.sin(omega*t)

# Numerical differentiation
vx = np.gradient(x, t)
vy = np.gradient(y, t)

ax = np.gradient(vx, t)
ay = np.gradient(vy, t)

# Magnitudes
speed = np.sqrt(vx**2 + vy**2)
acceleration = np.sqrt(ax**2 + ay**2)

# 1. Trajectory
plt.figure(figsize=(5, 5))
plt.plot(x, y)
plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.title("Trajectory")
plt.axis("equal")
plt.grid()
plt.show()

# 2. Velocity
plt.figure(figsize=(7, 4))
plt.plot(t, vx, label="vx")
plt.plot(t, vy, label="vy")
plt.plot(t, speed, label="speed")
plt.xlabel("time (s)")
plt.ylabel("velocity (m/s)")
plt.title("Velocity obtained from position data")
plt.grid()
plt.legend()
plt.show()

# 3. Acceleration
plt.figure(figsize=(7, 4))
plt.plot(t, ax, label="ax")
plt.plot(t, ay, label="ay")
plt.plot(t, acceleration, label="|a|")
plt.xlabel("time (s)")
plt.ylabel("acceleration (m/s²)")
plt.title("Acceleration obtained from velocity data")
plt.grid()
plt.legend()
plt.show()

print(f"Mean numerical speed = {np.mean(speed):.3f} m/s")
print(f"Theoretical speed R*omega = {R*omega:.3f} m/s")

print(f"Mean numerical acceleration = {np.mean(acceleration):.3f} m/s²")
print(f"Theoretical acceleration R*omega² = {R*omega**2:.3f} m/s²")
```
