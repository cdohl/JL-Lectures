In this experiment, we test how the sound from a loudspeaker decreases with distance.

### Measurement

Choose a starting distance $r$, for example

r=0.5\,\mathrm{m}.
$$

Measure the sound level at

$$
r,\qquad 2r,\qquad 4r,\qquad 8r.
$$

For $r=0.5$ m, these distances are

$$

$$

Record the values displayed by phyphox. These may be negative, for example $-20$ dB or $-35$ dB. This is fine: for this experiment we are interested in the **change** in sound level with distance, not its absolute value.

| distance $r$ (m) | phyphox level $L$ (dB) |
|---:|---:|
| | |

For an ideal point source, the sound intensity decreases according to

$$
I\propto\frac{1}{r^2}.
$$

Because the sound level in decibels is logarithmic, this leads to

$$
L=L_0-20\log_{10}(r),
$$

where $L_0$ is a constant that depends on the loudspeaker and the reference used by the microphone.

Therefore, every time the distance is doubled, we expect the measured level to decrease by approximately

$$
20\log_{10}(2)\approx6\,\mathrm{dB}.
$$

So ideally,

$$
L(2r)-L(r)\approx-6\,\mathrm{dB}.
$$

**c)** Why do we not need to know what $0$ dB means in phyphox for this experiment?

## 💻 8. Programming: Test the distance law

Now use Python to analyse your measurements.

For an ideal point source we expect

$$
L=L_0-20\log_{10}(r).
$$

This means that a plot of $L$ against $\log_{10}(r)$ should be a **straight line**.

The expected slope is

$$
-20\,\mathrm{dB/decade}.
$$

Notice that the measured dB values may be **negative**. We do **not** calculate their logarithm. The dB scale is already logarithmic.

Only the distance is transformed:

$$
x=\log_{10}(r).
$$

### Tasks

1. Enter your measured distances and phyphox dB values into the Python program below.
2. Plot $L$ as a function of distance $r$.
3. Plot $L$ as a function of $\log_{10}(r)$.
4. Fit a straight line to the second graph.
5. Compare your measured slope with the theoretical value

$$
-20\,\mathrm{dB/decade}.
$$

6. Does your experiment support the inverse-square law?

7. Calculate the change in dB for each doubling of distance. Is it approximately $-6$ dB?

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Enter YOUR measurements here
# ------------------------------------------------

# Distance from the loudspeaker in metres
r = np.array([0.5, 1.0, 2.0, 4.0])

# Sound level measured with phyphox in dB
# Replace these example values with your measurements
L = np.array([-18.0, -24.0, -30.0, -36.0])


# ------------------------------------------------
# 1. Plot sound level as a function of distance
# ------------------------------------------------

plt.figure(figsize=(7, 4))
plt.plot(r, L, "o-")

plt.xlabel("Distance r (m)")
plt.ylabel("phyphox level L (dB)")
plt.title("Sound level versus distance")
plt.grid()

plt.show()


# ------------------------------------------------
# 2. Logarithmic distance
# ------------------------------------------------

x = np.log10(r)

# Fit a straight line:
# L = slope * log10(r) + intercept

slope, intercept = np.polyfit(x, L, 1)

L_fit = slope * x + intercept


plt.figure(figsize=(7, 4))

plt.plot(x, L, "o", label="Measurement")
plt.plot(x, L_fit, "-", label="Linear fit")

plt.xlabel(r"$\log_{10}(r/\mathrm{m})$")
plt.ylabel("phyphox level L (dB)")
plt.title("Sound level versus logarithmic distance")

plt.grid()
plt.legend()

plt.show()


# ------------------------------------------------
# Results
# ------------------------------------------------

print(f"Measured slope: {slope:.1f} dB/decade")
print("Expected slope: -20 dB/decade")


# ------------------------------------------------
# 3. Change for every doubling of distance
# ------------------------------------------------

print("\nChange in sound level for each doubling:")

for i in range(len(L) - 1):
    change = L[i+1] - L[i]
    print(f"{r[i]:.2f} m -> {r[i+1]:.2f} m : {change:.1f} dB")

print("\nExpected change: approximately -6 dB per doubling")
```
