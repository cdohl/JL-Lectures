# 💻 7. Programming — determine an exponent from your experiment

Now use Python to analyse your own measurements.

Enter the **actual distances** you used and the corresponding values measured by phyphox.

For example, if your distances were 0.25 m, 0.50 m, 0.75 m and 1.00 m, replace the example values below with your own data.

```python
import numpy as np
import matplotlib.pyplot as plt

# Replace these with YOUR measured distances in metres
r = np.array([0.25, 0.50, 0.75, 1.00])

# Replace these with YOUR phyphox measurements
I = np.array([1.0, 0.6, 0.4, 0.3])

plt.figure(figsize=(7, 4))
plt.plot(r, I, "o-")
plt.xlabel("distance r (m)")
plt.ylabel("measured signal I")
plt.title("Measured sound signal")
plt.grid()
plt.show()
```

### Transform the data

If

$$
I=Cr^n,
$$

then

$$
\ln I=\ln C+n\ln r.
$$

Therefore a plot of $\ln I$ against $\ln r$ should approximately form a straight line.

Complete and run the next cell.

```python
log_r = np.log(r)
log_I = np.log(I)

plt.figure(figsize=(7, 4))
plt.plot(log_r, log_I, "o")
plt.xlabel("ln(r)")
plt.ylabel("ln(I)")
plt.title("Log-log representation of the measurements")
plt.grid()
plt.show()
```

### Estimate the exponent

The exponent $n$ is the slope of the graph.

You can estimate it by choosing two well-separated points and calculating

$$
n\approx
\frac{\ln I_2-\ln I_1}
{\ln r_2-\ln r_1}.
$$

Alternatively, the following Python command fits a straight line to all your measurements.

```python
n, intercept = np.polyfit(log_r, log_I, 1)

print("Estimated exponent n =", n)

fit = intercept + n * log_r

plt.figure(figsize=(7, 4))
plt.plot(log_r, log_I, "o", label="measurement")
plt.plot(log_r, fit, label=f"fit: slope = {n:.2f}")
plt.xlabel("ln(r)")
plt.ylabel("ln(I)")
plt.title("Determining a power-law exponent")
plt.grid()
plt.legend()
plt.show()
```
