# 2. Motion in one and three dimensions

## 2. 1. Position, Velocity, and Speed


* Description of the motion

To describe the motion of objects we need to introduce a system to refer to the object in space and time. For space this is a coordinate system. The motion is the described by the change of the location of the object.  Commonly in physics the location is indicated with the symbol $\vec{r(t)}$. As the objects move in space (sometimes 2 dimensions are sufficient, but most of the time in 3 dimensions), we have two or introduced a notation that accounts for this motion. This notation is termed vector. 


The velocity describes the change of the position vector with time, and the term acceleration the change of the velocity with time. 

 $$ \underbrace{\vec{r}(t)}_{\text{location as a function of time}} \leftrightarrow \underbrace{\vec{v}(t)}_{\text{velocity}} \leftrightarrow \underbrace{\vec{a}(t)}_{\text{acceleration}}$$

* point-like objects with mass $m$ 
  the total mass of the object is located at one point (zero volume), i.e. for planetary motion
  
* extended bodies (rigid body)
  system of point masses with fixed distances to each other

### 2.2 Definition of velocity and acceleration 

* $\vec{r}(t)$ ⇒ we focus first on one dimensional motion, i.e. a single component of the location vector changes with time $t$. Let us the x-axis for this motion, $x(t)$. Then we can plot the location $x$ as a function of time in a graph as below.
  
<img src="figs/3.2_1.png" alt="average velocity graph" style="width: 500px;"/>


$$ t_2 = t_1 +\Delta t \quad \Delta t=t_2-t_1$$

$$ x_2 = x_1 +\Delta x \quad \Delta x=x_2-x_1$$

**Average Velocity**

$$\bar{v}_x=\left< v_x \right> = \frac{x_2-x_1}{t_2-t_1}=\frac{\Delta x}{\Delta t}$$

What happens once we reduce the temporal interval $\Delta t$?

Please play with the program below and demonstrate that a reduction of the interval $[t_1,t_2]$ results in a better approximation of the slope.


```python
%matplotlib inline

import ipywidgets as widgets
import matplotlib.pyplot as plt
import matplotlib as mpl
import numpy as np
from IPython.display import display

mpl.rcParams['font.size'] = 15

# Data
t = np.linspace(0., 1., 1000)
v = t**3


def update_plot(t1t2):

    t1, t2 = t1t2
    x1 = t1**3
    x2 = t2**3

    # Create a new figure for this update
    fig, ax = plt.subplots(figsize=(8, 6))

    # Trajectory
    ax.plot(t, v)

    # Points
    ax.plot(t1, x1, 'ko')
    ax.plot(t2, x2, 'ko')

    # Secant
    ax.plot([t1, t2], [x1, x2], 'g')

    # Guide lines
    ax.plot([t1, t1], [0, x1], 'k-.')
    ax.plot([t2, t2], [0, x2], 'k-.')
    ax.plot([0, t1], [x1, x1], 'k-.')
    ax.plot([0, t2], [x2, x2], 'k-.')

    # Delta t arrows
    ax.arrow(
        t1, x1,
        t2-t1, 0,
        lw=1., fc='k', ec='k',
        head_width=.02, head_length=.02,
        length_includes_head=True
    )

    ax.arrow(
        t2, x1,
        t1-t2, 0,
        lw=1., fc='k', ec='k',
        head_width=.02, head_length=.02,
        length_includes_head=True
    )

    ax.text(
        t1 + (t2-t1)/2. - .04,
        x1 - .07,
        r'$\Delta t$'
    )

    # Delta x arrows
    ax.arrow(
        t2, x1,
        0, x2-x1,
        lw=1., fc='k', ec='k',
        head_width=.02, head_length=.02,
        length_includes_head=True
    )

    ax.arrow(
        t2, x2,
        0, x1-x2,
        lw=1., fc='k', ec='k',
        head_width=.02, head_length=.02,
        length_includes_head=True
    )

    ax.text(
        t2 + .02,
        x1 + (x2-x1)/2. - .04,
        r'$\Delta x$'
    )

    # Axes
    ax.set_xlabel(r'time $t$')
    ax.set_ylabel(r'location $x$')

    ax.set_xlim(0, 1.)
    ax.set_ylim(0, 1.)

    ax.set_xticks(
        [t1, t2],
        [r'$t_1$', r'$t_2$']
    )

    ax.set_yticks(
        [x1, x2],
        [r'$x_1$', r'$x_2$']
    )

    plt.show()
    plt.close(fig)


# Slider
tv = widgets.FloatRangeSlider(
    value=[.5, .8],
    min=.1,
    max=.9,
    step=.02,
    description=r'$[t_1,t_2]$:',
    orientation='horizontal',
    readout=True,
    readout_format='.1f',
    continuous_update=True
)

# Connect slider to plot
out = widgets.interactive_output(
    update_plot,
    {'t1t2': tv}
)

# Display
display(tv, out)
```

**Instantaneous Acceleration**

$$\lim_{\Delta t\rightarrow 0} \frac{\Delta v_x}{\Delta t} = \frac{\mathrm{d}v_x}{\mathrm{d}t}=a_x$$

$$a_x=\frac{\mathrm{d}^2 x}{\mathrm{d} t^2}$$

Derivatives of physical quantity with respect to time are conveniently abbreviated with a dot centered above the quantity. Thus, it reads here

$$v_x=\frac{\mathrm{d}x}{\mathrm{d}t}=\dot{x}$$
and
$$a_x=\frac{\mathrm{d}^2 x}{\mathrm{d}t^2}=\ddot{x}$$

### General one dimensional equation of motion for constant acceleration

$$x(t)=C_1 + C_2 t +C_3 t^2$$

$$v(t)= C_2 + 2 C_3 t$$

$$a = 2 C_3$$

where $C_1, C_2, C_3$ are real constants. The constants have the following context:

$$\begin{eqnarray}
  C_1 &=& x(t=0)=x_0 \, , \\
  C_2 &=& v(t=0)=v_0 \, , \\
  C_3 &=& \frac{1}{2} a\quad .\\
\end{eqnarray}$$

  
We can thus write above equation in the meaningful way:


$$x(t)=x_0 + v_0 t +\frac{1}{2}a t\tag{3.1.1}$$





### Free Fall

Let us use above equation to describe the motion of a ball dropped from a height $x$ at time $t=0$. Initially the ball is at rest. We have the freedom to choose the direction of $x$-direction.

<img src="figs/3.2_2.png" alt="free fall" style="width: 500px;"/>


The constant gravitational accleration in Magdeburg is 

$$a=g=9.81\,\mathrm{m/s}^2$$

With this knowledge we can solve for the distance/height of fall $h$, and the fall time for a distance/height $t$, i.e. 

$$h=\frac{1}{2}g t^2$$

$$t=\sqrt{2}\sqrt{\frac{h}{g}}$$


### Summary One-Dimensional Motion 

* at rest
  $$x=\mathrm{const.}, \quad v_x =0, \quad a_x = 0$$


* constant velocity
  $$x=x(t), \quad v_x = \mathrm{const.}, \quad a_x = 0$$


* constant acceleration 
  $$x=x(t), \quad v_x =v_x(t), \quad a_x = \mathrm{const.}$$


* non-constant acceleration
  $$x=x(t), \quad v_x =v_x(t), \quad a_x = a(t)$$


### 2.3 Position Vector in Cartesian Coordinate System 

* Three directions, which are pairwise orthogonal to each other

* These directions are along the unit vectors
  $$\vec{e}_x, \quad \vec{e}_y, \quad \vec{e}_z$$
  
  ⇒ Their orientation is given by the right hand rule for a right handed coordinate system.
  
  $$\vec{e}_x\,\underbrace{\times}_{\text{vector/cross product}}\,\vec{e}_y=\vec{e}_z$$
  
  The length of the unit vectors is always 1, the unit vector has *no* physical unit.
  
  $$\left|\vec{e}_x\right|=\left|\vec{e}_y\right|=\left|\vec{e}_z\right|=1$$
  
* Usage of the location vector:

* Location point to position $P1$ with coordinates $(x_1,y_1,z_1)$:
  $$\vec{r}_1=x_1\,\vec{e}_x+y_1\,\vec{e}_y+z_1\,\vec{e}_z$$
  
  Then the distance $r_1$ from the origin to the position is 
  
  $$\left|\vec{r}_1\right|=\sqrt{x_1^2+y_1^2+z_1^2}\quad .$$

* Arbitrary location vector
  $$\vec{r}=x\,\vec{e}_x+y\,\vec{e}_y+z\,\vec{e}_z$$
  
  please note that $x, y, z$ can be function of some phyiscal qunatities, e.g. time.
  
  
  <img src="figs/3.1_1.png" alt="position vector" style="width: 500px;"/>

Here are some examples of important vectorial physical quantities

* Velocity

  $$\vec{v}=v_x\,\vec{e}_x+v_y\,\vec{e}_y+v_z\,\vec{e}_z$$
  
* Acceleration

  $$\vec{a}=a_x\,\vec{e}_x+a_y\,\vec{e}_y+a_z\,\vec{e}_z$$

* Force

  $$\vec{F}=F_x\,\vec{e}_x+F_y\,\vec{e}_y+F_z\,\vec{e}_z$$

### 2.5 Compound motion: projectile trajectory

A compound motion is the undisturbed superimposition of multiple movements. Here we have an example of two undisturbed/uncorrelated motions, one parallel to $\vec{e}_x$ and one parallel to $\vec{e}_z$ that is named the projectile trajectory

The projectile motion starts at $t=0$ with an initial veloccity

$$
\begin{eqnarray} 
  v_{z,0}&=&v_0\,\sin\beta \\
  v_{x,0}&=&v_0\,\cos\beta \\
  v_0&=&\left|\vec{v}_0\right|
\end{eqnarray}
$$

 <img src="figs/3.4_1.png" width="260">
  

The resulting motion is shown in below figure:

```python
%matplotlib inline

import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
import matplotlib.patches as patches

mpl.rcParams['font.size'] = 15

# Parameters
g = 9.81
beta = 45.
v0 = 5.
H = 0.3

# Initial velocity components
vx0 = v0 * np.cos(beta * np.pi / 180.)
vz0 = v0 * np.sin(beta * np.pi / 180.)

# Horizontal coordinate
x = np.linspace(0, 2.1, 100)

# Projectile trajectory
z = -g/2 * x**2 / vx0**2 + vz0/vx0 * x + H

# Close old figures if cell is executed again
plt.close('all')

# Create figure
fig, ax = plt.subplots(figsize=(8, 4))

# Trajectory
ax.plot(x, z)

# Coordinate axes
ax.axhline(color='k', linewidth=1)
ax.axvline(color='k', linewidth=1)

ax.set_xlabel('x')
ax.set_ylabel('z')

ax.set_xticks([])
ax.set_yticks([])

# Hide frame
ax.spines['top'].set_visible(False)
ax.spines['right'].set_visible(False)
ax.spines['bottom'].set_visible(False)
ax.spines['left'].set_visible(False)

# Unit vectors
ax.arrow(
    0., 0.,
    .35, 0.,
    lw=1.,
    fc='k',
    ec='k',
    head_width=.05,
    head_length=.05,
    length_includes_head=True
)

ax.arrow(
    0., 0.,
    0., .25,
    lw=1.,
    fc='k',
    ec='k',
    head_width=.05,
    head_length=.05,
    length_includes_head=True
)

ax.text(.1, -.1, r'$\vec{e}_x$')
ax.text(-.1, .1, r'$\vec{e}_z$')


# Velocity vector
ss = .1

ax.arrow(
    0., H,
    vx0*ss, vz0*ss,
    lw=1.,
    fc='k',
    ec='k',
    head_width=.05,
    head_length=.05,
    length_includes_head=True
)

# Horizontal reference line for angle
ax.plot(
    [0, 1.5*vx0*ss],
    [H, H],
    'k--'
)

# Velocity label
ax.text(
    vx0*ss,
    H + vz0*ss*1.1,
    r'$\vec{v}_0$'
)


# Angle beta
arc = patches.Arc(
    (0., H),
    width=vx0*ss*1.5,
    height=vz0*ss*1.5,
    angle=0.,
    theta1=0.,
    theta2=np.arctan2(vz0, vx0) * 180./np.pi,
    linewidth=2.
)

ax.add_patch(arc)

# Beta label
ax.text(
    vx0*ss*.3,
    H + vz0*ss*.1,
    r'$\beta$'
)

plt.show()
```

Derivation of the projectile motion

The acceleration in $z$ and $x$-direction: 

$$a_z(t)=-g=\mathrm{const.}\quad a_x(t)=0\quad.$$

The initial velocity in $z$ and $x$-direction: 

$$v_z(t=0)=v_{z,0}\quad v_x(t=0)=v_{x,0}\quad.$$

The initial position:

$$z(t=0)=H\quad x(t=0)=0\quad.$$

Equation (3.1.1) describes a one-dimensional movement with uniform acceleration. We now enter the acceleration, initial velocity and initial position for the $x$ and $z$-directions separately:

Starting with the $z$-direction:

$$z(t)=-\frac{g}{2} t^2+v_{z,0}\,t+H\tag{3.4.1}$$

und now for the $x$-direction:

$$x(t)=v_{x,0}\,t\tag{3.4.2}$$

We can use equation (3.4.2) to remove the time from equation (3.4.1) and explicitly write down the trajectory $z(x)$. According to equation (3.4.2), $t=\frac{x}{v_{x,0}}$ and inserted into equation (3.4.1) we obtain the * parabola* of the projectile:

$$z(x)=-\frac{g}{2} \frac{x^2}{v_{x,0}^2}+\frac{v_{z,0}}{v_{x,0}}\,x+H\tag{3.4.3}$$


We can also express the initial velocities by $v_0$ and the angle $\beta$:

$$z(x)=-\frac{g}{2} \frac{x^2}{\left(v_{0}\,\cos\beta\right)^2}+\tan\beta\,x+H\tag{3.4.4}$$

Change the angle $\beta$ below and find the angle at which the throwing parabola reaches the minimum and maximum distance.

```python
%matplotlib inline

import ipywidgets as widgets
import matplotlib.pyplot as plt
import numpy as np
import matplotlib as mpl
import matplotlib.patches as patches
from IPython.display import display

mpl.rcParams['font.size'] = 15

# Parameters
g = 9.81
v0 = np.sqrt(2) * 2.5
H = 0.3


def update_plot(beta):

    # Convert angle to radians
    betarad = np.deg2rad(beta)

    # Initial velocity components
    vx0 = v0 * np.cos(betarad)
    vz0 = v0 * np.sin(betarad)

    # Horizontal coordinate
    x = np.linspace(0, 2.1, 1000)

    # Projectile trajectory
    z = -g/2 * x**2 / vx0**2 + vz0/vx0 * x + H

    # Only retain trajectory above the ground
    mask = z >= 0
    x = x[mask]
    z = z[mask]

    # Create fresh figure
    fig, ax = plt.subplots(figsize=(10, 6))

    # Trajectory
    ax.plot(x, z)

    # Coordinate axes
    ax.axhline(color='k', linewidth=1)
    ax.axvline(color='k', linewidth=1)

    ax.set_xlabel('x')
    ax.set_ylabel('z')

    ax.set_xticks([])
    ax.set_yticks([])

    # Hide frame
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['bottom'].set_visible(False)
    ax.spines['left'].set_visible(False)

    # Unit vectors
    ax.arrow(
        0., 0.,
        .35, 0.,
        lw=1.,
        fc='k',
        ec='k',
        head_width=.05,
        head_length=.05,
        length_includes_head=True
    )

    ax.arrow(
        0., 0.,
        0., .25,
        lw=1.,
        fc='k',
        ec='k',
        head_width=.05,
        head_length=.05,
        length_includes_head=True
    )

    ax.text(.1, -.1, r'$\vec{e}_x$')
    ax.text(-.1, .1, r'$\vec{e}_z$')

    # Velocity vector
    ss = .1

    ax.arrow(
        0., H,
        vx0*ss, vz0*ss,
        lw=1.,
        fc='k',
        ec='k',
        head_width=.05,
        head_length=.05,
        length_includes_head=True
    )

    # Horizontal reference line
    ax.plot(
        [0, 1.5*vx0*ss],
        [H, H],
        'k--'
    )

    # Velocity label
    ax.text(
        vx0*ss,
        H + vz0*ss*1.1,
        r'$\vec{v}_0$'
    )

    # Angle beta
    arc = patches.Arc(
        (0., H),
        width=vx0*ss*1.5,
        height=vz0*ss*1.5,
        angle=0.,
        theta1=0.,
        theta2=beta,
        linewidth=2.
    )

    ax.add_patch(arc)

    ax.text(
        vx0*ss*.3,
        H + vz0*ss*.1,
        r'$\beta$'
    )

    # Plot limits
    ax.set_xlim(-.1, 2.3)
    ax.set_ylim(-.1, 1.4)

    plt.show()

    # Important: release figure after displaying it
    plt.close(fig)


# Slider
beta_slider = widgets.FloatSlider(
    value=30.,
    min=5.,
    max=85.,
    step=5.,
    description=r'$\beta$:',
    continuous_update=True
)

# Connect slider to plot
out = widgets.interactive_output(
    update_plot,
    {'beta': beta_slider}
)

# Display slider and plot
display(beta_slider, out)
```

For the special case of $H=0$, we now determine the distance $x_w$. At impact, $z=0$ in equation (3.4.4) and we transform:.

$$x_x\,\tan\beta=\frac{g}{2}\frac{x_w^2}{v_0^2\,\cos^2\beta}$$

$$x_w=\frac{2\cos^2\beta\,v_0^2}{g}\,\tan\beta$$

$$x_w=\frac{2\cos\beta\,\sin\beta\,v_0^2}{g}$$

Insertion of a trigonometric identity results:

$$x_w=\frac{v_0^2}{g}\,\sin\left(2\beta\right)$$

The angle for the maximum throwing distance results from 

$$\frac{\mathrm{d}x_w}{\mathrm{d}\beta}=0$$ 

$$\frac{\mathrm{d}x_w}{\mathrm{d}\beta}=\frac{2\,v_0^2}{g}\,\cos\left(2\beta\right)=0$$

$$\cos 2\beta=0\quad\rightarrow\quad\beta=45^{\circ}$$
