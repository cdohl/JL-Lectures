# Mass, Amount, and Concentration

### Amount

To obtain the amount of substance we need to count the number of atoms or molecules

Unfortunately, these are typically a large number. We therefore use the unit for the amount of molecules *mole* that is abbreviated as 1 mol.

1 mol are $6.02214179\ \times 10^{23}$ entities (atomes or molecules).

With the help of the Avogadro constant $N_A=6.02214179\ \times 10^{23}$ mol$^{-1}$ we can easily convert from the number of moles to the number of molecules.

### Mass
To calculate the mass of 1 mol of any substance we look for the atomic number. Therefore look at the periodic table for the atomic mass number that is written below the element symbol in the periodic table, e.g. for Carbon it is 12.011. This means 1 mol of Carbon weighs 12.011 g.



Let us now calculate the mass of 1 mol or $6.02214179\ \times 10^{23}$ molecules of $\text{CO}_2$:

$(1 * 12.01 \text{g/mol} + 2 * 16.00 \text{g/mol}) * 1 \text{mol} =44.01\,$g.


### Concentration


The concentration of a substance can be given in various ways. Here we adapt the most common used in biology labs. It is 

\begin{equation}
    \text{concentration~} c=\frac{\text{amount of solute}~ n}{\text{volume of total solution}~ V} 
\end{equation}

To simplify writing we use the abbreviation of 1 M for 1 mol per liter for the molar concentration.  Let us now generate 1 M of table salt solution in water. For this we first weigh 1 mol of NaCl. This is 

$(1 *  22.99 \text{g/mol} + 1 * 35.45 \text{g/mol})* 1\,\text{mol}= 58.44\,$g

Then we add water to this weighted solid compound to reach a volume of 1 l. This way we have obtained a 1 M NaCl solution.

Let us now discuss this diagram and connect the nomenclature for $n$, $N$, $M$, $m$.

n: number of moles of a substance

N: number of entities of a substance, i.e. number of atoms or molecules

m: weight of a substance 

M: either molar concentration (mol/liter) or molar mass (g/mol)

The molar volume $V_m$ is used for gases, i.e. 1 mol of a gaseous substance at standard pressure and $T=0^\circ\,$degree occupies exactly 22.4 l, i.e. 22.4 l of $\text{CO}_2$ weigh 44.01 g.


 




We also include the term density, that will be covered later, yet it is a measure how much a certain volume of a substance weighs and is abbreviated with the $\rho$.

Let us now put the formulas together where we use $M$ for molar mass.

$$c=\frac{n}{V}$$

$$n=\frac{m}{M}$$

Here $m$ is mass, $n$ is amount of substance, and $V$ is the final solution volume. Convert masses to grams and volumes to litres before substituting.
