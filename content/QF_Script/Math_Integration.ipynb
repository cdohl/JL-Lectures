# Integration


Integration is the reverse operation of differentiation, but it also has an important graphical meaning:

> **An integral adds many small contributions together.**

For a positive function $f(x)$, the definite integral

$$
\int_a^b f(x)\,dx
$$

is the **area under the graph** between $x=a$ and $x=b$.

In biology, integration is useful whenever a **rate** is known and we want to find the accumulated amount.

# 1. Integration as an Area

Imagine dividing the interval from $a$ to $b$ into many narrow rectangles.

For one rectangle,

$$
\text{area}\approx f(x)\,\Delta x.
$$

Adding all rectangles gives

$$
\text{total area}\approx
\sum_i f(x_i)\Delta x.
$$

If the rectangles become narrower and narrower, the approximation approaches the integral:

$$
\boxed{
\int_a^b f(x)\,dx
}
$$

Use the slider below to increase the number of rectangles.

```python
%matplotlib inline

import numpy as np
import matplotlib.pyplot as plt
import ipywidgets as widgets
from IPython.display import display

def integration_demo(N=5):

    # Simple positive function
    f = lambda x: 0.5*x**2 + 1

    a = 0
    b = 3

    x = np.linspace(a, b, 400)

    # Rectangle width
    dx = (b-a)/N

    # Midpoints
    xm = a + (np.arange(N)+0.5)*dx
    heights = f(xm)

    approximation = np.sum(heights*dx)

    # Exact integral:
    # integral (0.5 x^2 + 1) dx = x^3/6 + x
    exact = b**3/6 + b - (a**3/6 + a)

    fig, ax = plt.subplots(figsize=(8,5.5))

    ax.plot(x, f(x), lw=2, label=r'$f(x)=\frac{1}{2}x^2+1$')

    ax.bar(
        xm, heights,
        width=dx,
        alpha=0.35,
        edgecolor='k',
        align='center'
    )

    ax.set_xlim(a, b)
    ax.set_ylim(0, 6)
    ax.set_xlabel('$x$')
    ax.set_ylabel('$f(x)$')
    ax.grid(alpha=0.2)

    info = (
        rf"$N={N}$ rectangles" "\n"
        rf"$\Delta x={dx:.3f}$" "\n\n"
        rf"rectangle sum $\approx {approximation:.3f}$" "\n"
        rf"exact integral $= {exact:.3f}$"
    )

    ax.text(
        0.03, 0.96, info,
        transform=ax.transAxes,
        va='top',
        bbox=dict(boxstyle='round', facecolor='white', alpha=0.9)
    )

    ax.legend(loc='lower right')

    plt.show()
    plt.close(fig)

N_slider = widgets.IntSlider(
    value=5,
    min=1,
    max=50,
    step=1,
    description='rectangles:',
    continuous_update=True
)

out = widgets.interactive_output(
    integration_demo,
    {'N': N_slider}
)

display(N_slider, out)
```

# 2. Integration and Differentiation

Integration is closely connected to differentiation.

If

$$
F'(x)=f(x),
$$

then $F(x)$ is called an **antiderivative** of $f(x)$.

We write

$$
\boxed{
\int f(x)\,dx=F(x)+C
}
$$

where $C$ is an arbitrary constant.

For example,

$$
\frac{d}{dx}x^3=3x^2.
$$

Therefore,

$$
\boxed{
\int 3x^2\,dx=x^3+C.
}
$$

Why do we need $C$?

Because

$$
\frac{d}{dx}(x^3+5)=3x^2
$$

and also

$$
\frac{d}{dx}(x^3-17)=3x^2.
$$

Differentiation removes constants, so integration cannot recover their value.

# 3. Rule 1 — The Power Rule

The most important integration rule is

$$
\boxed{
\int x^n\,dx
=
\frac{x^{n+1}}{n+1}+C
\qquad (n\neq-1)
}
$$

Compare this with differentiation:

$$
x^n
\quad\xrightarrow{\text{differentiate}}\quad
nx^{n-1}.
$$

For integration we essentially work backwards:

> **Increase the exponent by one and divide by the new exponent.**

Examples:

$$
\int x^2\,dx
=
\frac{x^3}{3}+C
$$

$$
\int x^4\,dx
=
\frac{x^5}{5}+C
$$

$$
\int \sqrt{x}\,dx
=
\int x^{1/2}\,dx
=
\frac{2}{3}x^{3/2}+C.
$$

# 4. Rule 2 — Constant Factors

A constant factor can remain in front of the integral:

$$
\boxed{
\int C f(x)\,dx
=
C\int f(x)\,dx
}
$$

Example:

$$
\int 4x^3\,dx
=
4\int x^3\,dx
=
4\frac{x^4}{4}+C
=
\boxed{x^4+C}.
$$

Another example:

$$
\int 6x^2\,dx
=
6\frac{x^3}{3}+C
=
\boxed{2x^3+C}.
$$

# 5. Rule 3 — Sums and Differences

Terms can be integrated separately:

$$
\boxed{
\int [f(x)+g(x)]\,dx
=
\int f(x)\,dx+\int g(x)\,dx
}
$$

and similarly for subtraction.

Example:

$$
\int(3x^2+2x-4)\,dx.
$$

Integrate term by term:

$$
3\frac{x^3}{3}
+
2\frac{x^2}{2}
-
4x+C.
$$

Therefore,

$$
\boxed{
\int(3x^2+2x-4)\,dx
=
x^3+x^2-4x+C.
}
$$

# 6. Rule 4 — Some Important Integrals

A few integrals are useful enough to remember:

| Function | Integral |
|---|---|
| $x^n$ | $\displaystyle\frac{x^{n+1}}{n+1}+C$ |
| $e^x$ | $e^x+C$ |
| $\frac{1}{x}$ | $\ln|x|+C$ |
| $\sin x$ | $-\cos x+C$ |
| $\cos x$ | $\sin x+C$ |

Notice the important special case

$$
\boxed{
\int\frac{1}{x}\,dx=\ln|x|+C.
}
$$

The ordinary power rule cannot be used for $x^{-1}$ because it would require division by zero.

# 7. Definite Integrals

A definite integral has limits:

$$
\int_a^b f(x)\,dx.
$$

If $F'(x)=f(x)$, then

$$
\boxed{
\int_a^b f(x)\,dx
=
F(b)-F(a).
}
$$

Example:

$$
\int_0^2 x^2\,dx.
$$

An antiderivative is

$$
F(x)=\frac{x^3}{3}.
$$

Therefore,

$$
\int_0^2 x^2\,dx
=
\left[\frac{x^3}{3}\right]_0^2
=
\frac{8}{3}-0
=
\boxed{\frac83}.
$$

For a positive function, this is the area under the graph between $0$ and $2$.

# 8. Substitution — Reversing the Chain Rule

Sometimes an integral contains a function **inside another function**.

Consider

$$
\int 2x(x^2+1)^3\,dx.
$$

The expression

$$
x^2+1
$$

appears inside the power, and its derivative

$$
2x
$$

also appears in the integral.

This suggests the substitution

$$
u=x^2+1.
$$

Then

$$
du=2x\,dx.
$$

The integral becomes

$$
\int u^3\,du.
$$

Now use the power rule:

$$
\int u^3\,du
=
\frac{u^4}{4}+C.
$$

Finally substitute back:

$$
\boxed{
\int 2x(x^2+1)^3\,dx
=
\frac{(x^2+1)^4}{4}+C.
}
$$

### A useful question

When looking at an integral, ask:

> **Can I identify an inner function and is its derivative also present?**

# 9. Integration by Parts — Reversing the Product Rule

Integration by parts is useful for products of functions.

The rule is

$$
\boxed{
\int u\,dv
=
uv-\int v\,du.
}
$$

A simple example is

$$
\int x e^x\,dx.
$$

Choose

$$
u=x
\qquad\text{and}\qquad
dv=e^x\,dx.
$$

Then

$$
du=dx
$$

and

$$
v=e^x.
$$

Insert these into the rule:

$$
\int xe^x\,dx
=
xe^x-\int e^x\,dx.
$$

Thus,

$$
\boxed{
\int xe^x\,dx
=
xe^x-e^x+C.
}
$$

or equivalently,

$$
\boxed{
\int xe^x\,dx
=
e^x(x-1)+C.
}
$$

For this introductory course, the important point is to recognize that integration by parts can simplify a **product**.

# 10. Biological Meaning: Integrating a Rate

Suppose $R(t)$ is the rate at which a substance is produced:

$$
R(t)=\frac{dM}{dt}.
$$

The amount produced between times $t_1$ and $t_2$ is

$$
\boxed{
\Delta M
=
\int_{t_1}^{t_2}R(t)\,dt.
}
$$

For example, suppose cells produce a substance at the rate

$$
R(t)=2t
\quad \mathrm{\mu g\,h^{-1}},
$$

where $t$ is measured in hours.

The amount produced during the first $3$ hours is

$$
\Delta M
=
\int_0^3
(2\,\mathrm{\mu g\,h^{-2}})\,t\,dt.
$$

Therefore,

$$
\Delta M
=
(1\,\mathrm{\mu g\,h^{-2}})
\left[t^2\right]_0^3
=
\boxed{9\,\mu\mathrm{g}}.
$$

Notice how the units also make sense:

$$
\mathrm{\frac{\mu g}{h}}\times\mathrm{h}
=
\mathrm{\mu g}.
$$

Integration converts a **rate** into an **accumulated amount**.

# 11. Integration Cheat Sheet

### Basic rules

$$
\boxed{
\int x^n\,dx
=
\frac{x^{n+1}}{n+1}+C
\qquad(n\neq-1)
}
$$

$$
\boxed{
\int C f(x)\,dx=C\int f(x)\,dx
}
$$

$$
\boxed{
\int(f+g)\,dx
=
\int f\,dx+\int g\,dx
}
$$

### Important functions

$$
\int e^x\,dx=e^x+C
$$

$$
\int\frac1x\,dx=\ln|x|+C
$$

$$
\int\sin x\,dx=-\cos x+C
$$

$$
\int\cos x\,dx=\sin x+C
$$

### Definite integral

$$
\boxed{
\int_a^b f(x)\,dx=F(b)-F(a)
}
$$

### Substitution

Look for an inner function and its derivative:

$$
u=g(x),\qquad du=g'(x)\,dx.
$$

### Integration by parts

$$
\boxed{
\int u\,dv=uv-\int v\,du
}
$$
