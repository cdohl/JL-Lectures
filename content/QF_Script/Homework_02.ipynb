## 8. 💻 Programming — crossing a river

A swimmer moves with speed

$$
v_s=1.2\,\mathrm{m/s}
$$

relative to the water. The river flows in the $x$-direction with speed

$$
v_r=0.8\,\mathrm{m/s}.
$$

The river is

$$
50\,\mathrm{m}
$$

wide.

The swimmer can choose the angle $\alpha$ at which to swim. We measure $\alpha$ from the direction straight across the river; positive angles point upstream.

The velocity relative to the ground is the vector sum

$$
\vec v_{\mathrm{ground}}
=
\vec v_{\mathrm{swimmer}}
+
\vec v_{\mathrm{river}}.
$$

### Tasks

**a)** First set

$$
\alpha=0^\circ.
$$

Run the program. How long does the swimmer need to cross the river, and how far downstream does the swimmer arrive?

**b)** Now change `alpha_deg`. Find approximately the angle for which the swimmer arrives directly opposite the starting point, i.e. at

$$
x=0.
$$

**c)** Calculate this angle analytically using the velocity vectors. Compare your result with the value found using Python.

**d)** Increase the river speed `vr`. How does the required swimming angle change?

**e)** What happens if the river becomes faster than the swimmer,

$$
v_r>v_s?
$$

Can you still find an angle that makes the swimmer arrive directly opposite the starting point?

**f)** Explain your observation using vectors. Derive the condition that must be satisfied for the swimmer to cancel the downstream velocity completely.

> The purpose of the program is not just to calculate one trajectory. Use it as a numerical experiment: change one parameter, observe what happens, and then explain the result mathematically.

```python
import numpy as np
import matplotlib.pyplot as plt

# ------------------------------------------------
# Parameters
# ------------------------------------------------

vs = 1.2       # swimmer speed relative to water (m/s)
vr = 0.8       # river speed (m/s)
width = 50.0   # river width (m)

# Angle measured from straight across the river.
# Positive angles point upstream.
alpha_deg = 0.0

alpha = np.deg2rad(alpha_deg)

# ------------------------------------------------
# Velocity relative to the ground
# ------------------------------------------------

# x: along the river, positive downstream
# y: across the river

vx = vr - vs * np.sin(alpha)
vy = vs * np.cos(alpha)

# Time needed to reach the opposite bank
t_cross = width / vy

# Trajectory
t = np.linspace(0, t_cross, 100)

x = vx * t
y = vy * t

# ------------------------------------------------
# Plot
# ------------------------------------------------

plt.figure(figsize=(6, 5))

plt.plot(x, y)
plt.scatter([0, x[-1]], [0, y[-1]])

plt.xlabel("Position along river x (m)")
plt.ylabel("Position across river y (m)")
plt.title(f"River crossing: angle = {alpha_deg:.1f}°")

plt.grid()
plt.show()

# ------------------------------------------------
# Results
# ------------------------------------------------

print(f"Velocity along river vx = {vx:.2f} m/s")
print(f"Velocity across river vy = {vy:.2f} m/s")
print(f"Crossing time = {t_cross:.1f} s")
print(f"Downstream displacement = {x[-1]:.1f} m")
```
